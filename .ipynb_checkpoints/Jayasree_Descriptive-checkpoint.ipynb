## 1. How elderly is this patient population, and how common is underweight/malnutrition risk, especially among the oldest patients?
### Reasoning:  
Heart failure elderly patients often face malnutrition/cachexia, especially being underweight. So we first check: is the cohort elderly, and is underweight common.

### Metrics used:

Age category – shows risk across age groups

Gender – checks if risk differs by sex

BMI – basic nutrition screening

Underweight flag (BMI < 18.5) – clear risk indicator

### Why these:  
They’re simple, available for all patients, and good first-pass screens for malnutrition risk.

```python
import pandas as pd

df_demography = pd.read_csv("Team05_PyCraft_Cleaned_data/demography_cleaned.csv")
age_order = ['21-29','29-39','39-49','49-59','59-69','69-79','79-89','89-110']

# ============================================================
# KPI GROUP 1: Population Overview
# ============================================================
total_patients = len(df_demography)
pct_female = round((df_demography['gender'] == 'Female').mean() * 100, 1)
pct_male = round((df_demography['gender'] == 'Male').mean() * 100, 1)
pct_elderly = round(df_demography['age_category'].isin(['69-79','79-89','89-110']).mean() * 100, 1)

print("=== KPI 1: POPULATION OVERVIEW ===")
print(f"Total patients:        {total_patients}")
print(f"% Female:              {pct_female}%")
print(f"% Male:                {pct_male}%")
print(f"% Elderly (69+):       {pct_elderly}%")
print()

# ============================================================
# KPI GROUP 2: Age & Gender Distribution (detail table)
# ============================================================
print("=== KPI 2: AGE & GENDER BREAKDOWN ===")
age_gender_table = pd.crosstab(df_demography['age_category'], df_demography['gender']).reindex(age_order)
print(age_gender_table)
print()

# ============================================================
# KPI GROUP 3: Nutritional Risk (BMI-based)
# ============================================================
df_demography['underweight'] = df_demography['bmi'] < 18.5

avg_bmi = round(df_demography['bmi'].mean(), 1)
pct_underweight_overall = round(df_demography['underweight'].mean() * 100, 1)
pct_underweight_oldest = round(
    df_demography[df_demography['age_category'] == '89-110']['underweight'].mean() * 100, 1
)

print("=== KPI 3: NUTRITIONAL RISK (BMI) ===")
print(f"Average BMI (overall):           {avg_bmi} kg/m²")
print(f"% Underweight (overall):         {pct_underweight_overall}%")
print(f"% Underweight (age 89-110):      {pct_underweight_oldest}%")
print()

print("BMI trend by age category:")
print(df_demography.groupby('age_category')['bmi'].mean().reindex(age_order).round(2))
print()

print("% Underweight by age category:")
print((df_demography.groupby('age_category')['underweight'].mean() * 100).reindex(age_order).round(1))
```

```python
import matplotlib.pyplot as plt

age_order = ['21-29','29-39','39-49','49-59','59-69','69-79','79-89','89-110']

# --- Chart 1: Age & Gender distribution with labels ---
age_gender_table = pd.crosstab(df_demography['age_category'], df_demography['gender']).reindex(age_order)

fig, ax = plt.subplots(figsize=(10, 5))
age_gender_table.plot(kind='bar', ax=ax)

for container in ax.containers:
    ax.bar_label(container, fontsize=8)

plt.title("Age & Gender Distribution")
plt.xlabel("Age Category")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# --- Chart 2: % Underweight by age category with labels ---
underweight_by_age = (df_demography.groupby('age_category')['underweight'].mean() * 100).reindex(age_order)

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(underweight_by_age.index, underweight_by_age.values, color='orange')

ax.bar_label(bars, fmt='%.1f%%', fontsize=9)

plt.title("% of Patients Underweight (BMI < 18.5) by Age Category")
plt.xlabel("Age Category")
plt.ylabel("% Underweight")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Insights:

72.8% are 69+ → truly elderly cohort

25.2% underweight overall

36.6% underweight in 89+ → strong age-linked trend

### How it helps:

Shows underweight is a major subgroup, not a minor issue

Justifies using BMI/underweight in predictive models

Supports targeted interventions (e.g., nutrition screening for 79+)

Confirms data cleaning worked (smooth age trend)

## 2. What proportion of patients died during hospitalization vs. were discharged home, and how does that break down by admission ward (ICU, Cardiology, General)? (hospitalization)

### Reasoning

Mortality is the clearest outcome measure — comparing it by ward shows whether sicker patients are actually being placed where they'd get closest monitoring.

What we measured

hospitalization_outcome (Alive/Dead/DischargeAgainstOrder) × admission_ward

```python
import pandas as pd

df_hospitalization = pd.read_csv("Team05_PyCraft_Cleaned_data/hospitalization_discharge_cleaned.csv")

# ============================================================
# KPI GROUP 1: Overall Discharge Outcome
# ============================================================
total_patients = len(df_hospitalization)
pct_died = round((df_hospitalization['hospitalization_outcome'] == 'Dead').mean() * 100, 2)
pct_alive = round((df_hospitalization['hospitalization_outcome'] == 'Alive').mean() * 100, 2)
pct_against_order = round((df_hospitalization['hospitalization_outcome'] == 'DischargeAgainstOrder').mean() * 100, 2)

print("=== KPI 1: OVERALL DISCHARGE OUTCOME ===")
print(f"Total patients:              {total_patients}")
print(f"% Died in-hospital:          {pct_died}%")
print(f"% Discharged alive:          {pct_alive}%")
print(f"% Discharged against order:  {pct_against_order}%")
print()

# ============================================================
# KPI GROUP 2: Outcome by Admission Ward
# ============================================================
print("=== KPI 2: OUTCOME BY ADMISSION WARD ===")
ward_counts = df_hospitalization['admission_ward'].value_counts()
print("Patients per ward:")
print(ward_counts)
print()

outcome_by_ward = (
    pd.crosstab(df_hospitalization['admission_ward'], df_hospitalization['hospitalization_outcome'], normalize='index') * 100
).round(1)
print("Outcome % by ward:")
print(outcome_by_ward)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))
outcome_by_ward.plot(kind='bar', ax=ax)

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', fontsize=8)

plt.title("Discharge Outcome by Admission Ward")
plt.xlabel("Admission Ward")
plt.ylabel("% of Patients")
plt.xticks(rotation=0)
plt.legend(title="Outcome")
plt.tight_layout()
plt.show()
```

## Insights

Key numbers
Overall in-hospital mortality: 0.55%
ICU mortality: 6.7% (~12x the overall rate)
ICU "discharge against order": 13.3% (highest of all wards)

Insight

Mortality is heavily concentrated in ICU — confirms ward placement reflects real severity, and makes admission_ward a strong feature for later prediction. Caveat: ICU has only 15 patients, so 6.7% = just 1 death — report as "1 of 15," not as a precise rate.


## 3. Which heart failure type is most dangerous
Reasoning:  
Heart failure isn’t one single disease — left, right, and biventricular failure behave differently. Since your cohort is elderly and comorbidity‑heavy, it’s important to check whether the type of heart failure itself is an independent risk factor for death or readmission.

Metrics used:

Heart failure type (Left / Right / Both)

Mortality at 6 months

Readmission at 6 months

LVEF (heart pumping efficiency)

Why these:  
They let you check whether worse outcomes in “Both” are due to true disease mechanism or simply lower pumping efficiency.

```python
import pandas as pd

df_cardiac_complications = pd.read_csv("Team05_PyCraft_Cleaned_data/cardiac_cleaned.csv")
df_hospitalization = pd.read_csv("Team05_PyCraft_Cleaned_data/hospitalization_discharge_cleaned.csv")

merged = df_cardiac_complications.merge(
    df_hospitalization[['patient_id', 'death_6m', 'readmission_6m', 'hospitalization_outcome']],
    on='patient_id', how='left'
)

# ============================================================
# KPI: Risk by Heart Failure Type
# ============================================================
print("=== Patients per heart failure type ===")
print(merged['heart_failure_type'].value_counts())
print()

print("=== 6-month mortality by type ===")
mortality_by_type = (merged.groupby('heart_failure_type')['death_6m'].mean() * 100).round(2)
print(mortality_by_type)
print()

print("=== 6-month readmission by type ===")
readmission_by_type = (merged.groupby('heart_failure_type')['readmission_6m'].mean() * 100).round(2)
print(readmission_by_type)
print()

print("=== Average pumping efficiency (LVEF) by type ===")
print(merged.groupby('heart_failure_type')['heart_pumping_efficiency'].mean().round(1))
```

```python
import matplotlib.pyplot as plt

hf_types = merged['heart_failure_type'].dropna().unique().tolist()
mortality_vals = mortality_by_type.reindex(hf_types)
readmission_vals = readmission_by_type.reindex(hf_types)

x = range(len(hf_types))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
bars1 = ax.bar([i - width/2 for i in x], mortality_vals, width, label='6-Month Mortality %')
bars2 = ax.bar([i + width/2 for i in x], readmission_vals, width, label='6-Month Readmission %')

ax.bar_label(bars1, fmt='%.1f%%', fontsize=8)
ax.bar_label(bars2, fmt='%.1f%%', fontsize=8)

ax.set_xticks(x)
ax.set_xticklabels(hf_types)
ax.set_xlabel("Heart Failure Type")
ax.set_ylabel("Percent of Patients")
ax.set_title("Risk by Heart Failure Type")
ax.legend()
plt.tight_layout()
plt.show()
```

## Insights
Biventricular (“Both”) is the riskiest: highest mortality (3.11%), highest readmission (42.2%), and most common (74%).

Right‑sided shows 0% mortality but has only 51 patients, so the sample is too small to interpret.

LVEF is similar across all types (49–53%), meaning the higher risk in “Both” isn’t just due to worse pumping — the combined left+right involvement adds extra risk.

### How it helps analysis:

Identifies a high‑risk clinical subgroup, not just a demographic one.

Justifies using heart_failure_type as a predictive model feature.

Shows the risk isn’t explained by LVEF alone → the type itself matters.

Sets up next steps: evaluate whether biventricular patients receive appropriately intensive monitoring or treatment.

## 4. How severe is heart failure at admission across genders, based on heart_failure_severity (NYHA class) and heart_damage_level (Killip grade)?

### Reasoning:
Heart failure severity at admission is measured on two different clinical scales, and they capture different things:

- **heart_failure_severity (NYHA class)** — how much heart failure limits daily activity: class 2 = mild limitation, class 3 = marked limitation, class 4 = symptoms even at rest.
- **heart_damage_level (Killip grade)** — physical signs of acute heart failure on examination: grade 1 = no congestion, grade 2 = mild congestion (rales), grade 3 = pulmonary oedema, grade 4 = cardiogenic shock.

Our cohort is 57.9% female (Question 1), and men and women often present with heart failure differently. Before building risk models we need to know whether one sex arrives sicker, so that gender is not mistaken for a risk factor when it is really a proxy for severity.

### Metrics used:

Row percentages — share of each sex in every NYHA class and every Killip grade (fair comparison, since there are more women than men)

Share in the most severe category — NYHA class 4 (symptoms at rest) and Killip grade 4 (cardiogenic shock)

Chi-square test of independence — is the severity distribution associated with sex, or could the difference be chance?

Cramér's V — effect size: how *large* the difference is (0.1 = small, 0.3 = medium, 0.5 = large)

### Why these:
Row percentages put both sexes on the same 0–100% scale, the chi-square test tells us whether a difference is real, and Cramér's V tells us whether it is big enough to matter clinically. Both scales have no missing values, so every patient is included.

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

df_demography = pd.read_csv("Team05_PyCraft_Cleaned_data/demography_cleaned.csv")
df_cardiac_complications = pd.read_csv("Team05_PyCraft_Cleaned_data/cardiac_cleaned.csv")

severity = df_demography[['patient_id', 'gender']].merge(
    df_cardiac_complications[['patient_id', 'heart_failure_severity', 'heart_damage_level']],
    on='patient_id', how='inner', validate='one_to_one'
)

# ============================================================
# KPI GROUP 1: Data coverage
# ============================================================
print("=== KPI 1: DATA COVERAGE ===")
print(f"Patients analysed:                     {len(severity)}")
print(f"Missing heart_failure_severity (NYHA): {severity['heart_failure_severity'].isna().sum()}")
print(f"Missing heart_damage_level (Killip):   {severity['heart_damage_level'].isna().sum()}")
print(severity['gender'].value_counts().to_string())
print()

# ============================================================
# KPI GROUP 2: Severity distribution by gender (row %)
# ============================================================
nyha_counts = pd.crosstab(severity['gender'], severity['heart_failure_severity'])
killip_counts = pd.crosstab(severity['gender'], severity['heart_damage_level'])
nyha_pct = (pd.crosstab(severity['gender'], severity['heart_failure_severity'], normalize='index') * 100).round(1)
killip_pct = (pd.crosstab(severity['gender'], severity['heart_damage_level'], normalize='index') * 100).round(1)

print("=== KPI 2a: NYHA CLASS (heart_failure_severity) BY GENDER, % of each sex ===")
print(nyha_pct)
print()
print("=== KPI 2b: KILLIP GRADE (heart_damage_level) BY GENDER, % of each sex ===")
print(killip_pct)
print()

# ============================================================
# KPI GROUP 3: Most severe category + statistical significance
# ============================================================
def chi_square_with_effect_size(counts):
    chi2, p, dof, _ = chi2_contingency(counts)
    cramers_v = np.sqrt(chi2 / (counts.values.sum() * (min(counts.shape) - 1)))
    return chi2, p, cramers_v

chi2_nyha, p_nyha, v_nyha = chi_square_with_effect_size(nyha_counts)
chi2_killip, p_killip, v_killip = chi_square_with_effect_size(killip_counts)

print("=== KPI 3: MOST SEVERE CATEGORY & SIGNIFICANCE ===")
print(f"NYHA class 4 (symptoms at rest):   Female {nyha_pct.loc['Female', 4]}% | Male {nyha_pct.loc['Male', 4]}%")
print(f"Killip grade 4 (cardiogenic shock): Female {killip_pct.loc['Female', 4]}% | Male {killip_pct.loc['Male', 4]}%")
print(f"NYHA vs gender:   chi2 = {chi2_nyha:.2f}, p = {p_nyha:.4f}, Cramer's V = {v_nyha:.3f}")
print(f"Killip vs gender: chi2 = {chi2_killip:.2f}, p = {p_killip:.4f}, Cramer's V = {v_killip:.3f}")
```

### Visual:
One chart shows both severity scales for both sexes. Each bar is one sex on one scale and adds up to 100% of that sex's patients. Colours run from **green (mildest)** to **dark red (most severe)**, so a bar with more red on its right-hand end means that sex arrives sicker.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Bars: NYHA block on top, Killip block below, one bar per sex
rows = [('heart_failure_severity', 'Female'), ('heart_failure_severity', 'Male'),
        ('heart_damage_level', 'Female'), ('heart_damage_level', 'Male')]
y_pos = [3.55, 2.85, 1.25, 0.55]
level_names = {'heart_failure_severity': {2: 'Class 2', 3: 'Class 3', 4: 'Class 4'},
               'heart_damage_level': {1: 'Grade 1', 2: 'Grade 2', 3: 'Grade 3', 4: 'Grade 4'}}
level_colours = {'heart_failure_severity': {2: '#7BC47F', 3: '#F2C14E', 4: '#B3261E'},
                 'heart_damage_level': {1: '#7BC47F', 2: '#F2C14E', 3: '#E8833A', 4: '#B3261E'}}
pct_tables = {'heart_failure_severity': nyha_pct, 'heart_damage_level': killip_pct}

fig, ax = plt.subplots(figsize=(12, 6))
for (scale, sex), y in zip(rows, y_pos):
    left = 0
    for level, share in pct_tables[scale].loc[sex].items():
        ax.barh(y, share, left=left, height=0.65, color=level_colours[scale][level], edgecolor='white')
        text_colour = 'white' if level_colours[scale][level] == '#B3261E' else 'black'
        if share >= 6:   # label inside the segment when there is room
            ax.text(left + share / 2, y, f"{level_names[scale][level]}\n{share:.1f}%", ha='center', va='center', fontsize=8.5, color=text_colour)
        left += share
    # Call out the most severe category at the end of each bar
    worst = pct_tables[scale].loc[sex].iloc[-1]
    label = 'symptoms at rest' if scale == 'heart_failure_severity' else 'cardiogenic shock'
    ax.text(101.5, y, f"most severe ({label}): {worst:.1f}%", va='center', fontsize=9,
            fontweight='bold', color='#B3261E')

ax.set_yticks(y_pos)
ax.set_yticklabels(['Female', 'Male', 'Female', 'Male'], fontsize=10)
# Block headings: scale name + significance test
ax.text(0, 4.05, f"NYHA class (heart_failure_severity)   ·   chi-square p = {p_nyha:.3f}, Cramer's V = {v_nyha:.2f}",
        fontsize=10, fontweight='bold', va='center')
ax.text(0, 1.75, f"Killip grade (heart_damage_level)   ·   chi-square p = {p_killip:.3f}, Cramer's V = {v_killip:.2f}",
        fontsize=10, fontweight='bold', va='center')
ax.axhline(2.15, color='grey', lw=0.8, ls='--')
ax.set_ylim(0.1, 4.3)
ax.set_xlim(0, 100)
ax.set_xlabel("% of patients within each sex (each bar = 100%)")
ax.set_title("Heart failure severity at admission by gender: NYHA class and Killip grade", fontsize=13, pad=12)
ax.legend(handles=[Patch(color='#7BC47F', label='Mildest level'), Patch(color='#F2C14E', label='Moderate'),
                   Patch(color='#E8833A', label='Severe (Killip grade 3: pulmonary oedema)'), Patch(color='#B3261E', label='Most severe level')],
          loc='upper center', bbox_to_anchor=(0.5, -0.12), ncol=4, fontsize=8.5, frameon=False)
for side in ['top', 'right']:
    ax.spines[side].set_visible(False)
plt.tight_layout()
plt.show()
```

## Insights

**Both sexes mostly arrive with advanced heart failure:** 80.6% of women and 85.0% of men are NYHA class 3–4, and the single most common profile for both is NYHA class 3 (about 52% and 51%) with Killip grade 2 (about 51% and 52%).

**Men arrive slightly sicker on both scales:**
- **NYHA class 4 (symptoms at rest):** 34.0% of men vs 28.3% of women. Women are more often in the mildest class 2 (19.4% vs 15.0%).
- **Killip grade 4 (cardiogenic shock):** 4.0% of men vs 2.2% of women, almost double. Women are more often in grade 1, with no congestion (28.0% vs 23.8%).
- **Killip grade 3 (pulmonary oedema):** almost identical (19.3% vs 19.9%), so the extra severity in men is concentrated in shock.

**The differences are real but small.** Both are statistically significant (NYHA p = 0.005, Killip p = 0.029), but the effect sizes are small (Cramér's V ≈ 0.07). Sex shifts severity slightly; it does not define it.

**Caveat:** Killip grade 4 is rare (60 patients: 26 women, 34 men), so the 4.0% vs 2.2% gap rests on small numbers.

### How it helps analysis:

Shows that men start from a slightly worse position, so any higher mortality in men later should be checked against admission severity before blaming sex itself.

Supports using heart_failure_severity and heart_damage_level together with gender (or adjusting for them) in predictive models.

Highlights cardiogenic shock (Killip grade 4) as the category where the sex gap is largest — a useful early-warning flag for triage.

Confirms the cleaned severity columns are complete (no missing values), so they can be used for every patient.

## 5. How does the risk of death change over the first 4 weeks after admission, depending on the patient's consciousness level at admission?

### Reasoning:
Impaired consciousness in heart failure usually means the brain is not getting enough blood or oxygen (for example in cardiogenic shock), which is a warning sign of imminent death. A single 28-day death rate hides *when* patients die. Tracking cumulative mortality day by day for each consciousness level shows how steep the danger is and how early it strikes, which decides how quickly clinicians must act.

`consciousness` has four levels, from least to most severe: Clear → ResponsiveToSound → ResponsiveToPain → Nonresponsive.

### Metrics used:

Consciousness level at admission (responsivenes_cleaned.csv)

death_time_days — day of death counted from admission (hospitalization_discharge_cleaned.csv)

Cumulative mortality at the end of week 1, 2, 3 and 4 (day 7, 14, 21, 28) per consciousness level

death_28d flag — used to cross-check the day-by-day counts

Relative risk and Fisher's exact test — impaired (any level other than Clear) vs Clear; Fisher's test is used because the impaired groups are small

### Why these:
death_time_days is recorded for every patient who died within 28 days, so the curve can be built day by day without guessing. Cumulative percentages turn that into an easy-to-read risk gradient over time.

```python
import pandas as pd
import numpy as np
from scipy.stats import fisher_exact

df_responsivenes = pd.read_csv("Team05_PyCraft_Cleaned_data/responsivenes_cleaned.csv")
df_hospitalization = pd.read_csv("Team05_PyCraft_Cleaned_data/hospitalization_discharge_cleaned.csv")

consciousness_order = ['Clear', 'ResponsiveToSound', 'ResponsiveToPain', 'Nonresponsive']
mort = df_responsivenes[['patient_id', 'consciousness', 'gcs']].merge(
    df_hospitalization[['patient_id', 'death_28d', 'death_time_days']],
    on='patient_id', how='inner', validate='one_to_one'
)

# ============================================================
# KPI GROUP 1: Data check — does the day of death agree with the 28-day flag?
# ============================================================
died_by_day_28 = mort['death_time_days'] <= 28
print("=== KPI 1: DATA CHECK ===")
print(f"Patients:                                   {len(mort)}")
print(f"Deaths within 28 days (death_28d flag):     {mort['death_28d'].sum()}")
print(f"Deaths with death_time_days <= 28:          {died_by_day_28.sum()}")
print(f"28-day deaths missing a day of death:       {(mort['death_28d'].eq(1) & mort['death_time_days'].isna()).sum()}")
print("Flag and day of death agree for every patient:", bool((died_by_day_28.astype(int) == mort['death_28d']).all()))
print()

# ============================================================
# KPI GROUP 2: Cumulative mortality at the end of each week
# ============================================================
group_size = mort['consciousness'].value_counts().reindex(consciousness_order)
weekly = pd.DataFrame(index=consciousness_order)
for week, day in zip([1, 2, 3, 4], [7, 14, 21, 28]):
    deaths = mort[mort['death_time_days'] <= day].groupby('consciousness').size().reindex(consciousness_order, fill_value=0)
    weekly[f'Week {week} (day {day})'] = (deaths / group_size * 100).round(1)
weekly.insert(0, 'Patients', group_size)

print("=== KPI 2: CUMULATIVE MORTALITY (%) BY CONSCIOUSNESS LEVEL ===")
print(weekly.to_string())
print()

# ============================================================
# KPI GROUP 3: Impaired vs Clear — how much more dangerous?
# ============================================================
mort['impaired'] = mort['consciousness'] != 'Clear'
risk_impaired = mort.loc[mort['impaired'], 'death_28d'].mean() * 100
risk_clear = mort.loc[~mort['impaired'], 'death_28d'].mean() * 100
odds_ratio, p_fisher = fisher_exact(pd.crosstab(mort['impaired'], mort['death_28d']))
share_of_deaths = mort.loc[mort['impaired'], 'death_28d'].sum() / mort['death_28d'].sum() * 100
week1_share = (mort['death_time_days'] <= 7).sum() / mort['death_28d'].sum() * 100

print("=== KPI 3: IMPAIRED vs CLEAR CONSCIOUSNESS (28 DAYS) ===")
print(f"Impaired patients:                  {mort['impaired'].sum()} ({mort['impaired'].mean() * 100:.1f}% of all patients)")
print(f"28-day mortality, impaired:         {risk_impaired:.1f}%")
print(f"28-day mortality, clear:            {risk_clear:.1f}%")
print(f"Relative risk:                      {risk_impaired / risk_clear:.1f}x")
print(f"Fisher's exact test p-value:        {p_fisher:.2e}")
print(f"Share of all 28-day deaths:         {share_of_deaths:.1f}%")
print(f"Share of 28-day deaths in week 1:   {week1_share:.1f}%")
```

### Visual:
The chart shows cumulative mortality day by day for the first 28 days, one line per consciousness level. The steeper and higher a line climbs, the more dangerous that level is. Dotted lines mark the end of each week, and the labels on the right give the final 28-day rate with the number of patients in each group.

```python
import matplotlib.pyplot as plt

colours = {'Clear': '#4C72B0', 'ResponsiveToSound': '#DD8452', 'ResponsiveToPain': '#C44E52', 'Nonresponsive': '#8C0000'}
days = np.arange(0, 29)

fig, ax = plt.subplots(figsize=(11, 6))
for level in consciousness_order:
    group = mort[mort['consciousness'] == level]
    cumulative = [(group['death_time_days'] <= d).mean() * 100 for d in days]
    ax.step(days, cumulative, where='post', color=colours[level], lw=2.5, label=f"{level} (n = {len(group)})")
    ax.plot([7, 14, 21, 28], [cumulative[7], cumulative[14], cumulative[21], cumulative[28]], 'o', color=colours[level], ms=6)
    ax.text(28.6, cumulative[28], f"{cumulative[28]:.1f}%", va='center', fontsize=10, fontweight='bold', color=colours[level])

for d in [7, 14, 21, 28]:
    ax.axvline(d, color='grey', ls=':', lw=1)
ax.set_xticks([0, 7, 14, 21, 28])
ax.set_xticklabels(['Admission', 'Week 1', 'Week 2', 'Week 3', 'Week 4'])
ax.set_xlim(0, 31.5)
ax.set_ylim(0, 100)
ax.set_ylabel("Cumulative mortality (% of patients in the group)")
ax.set_title("Mortality gradient over the first 4 weeks by consciousness level at admission", fontsize=13)
ax.legend(title="Consciousness at admission", loc='center right', bbox_to_anchor=(0.97, 0.45), fontsize=9)
for side in ['top', 'right']:
    ax.spines[side].set_visible(False)
plt.tight_layout()
plt.show()
```

## Insights

**A steep gradient.** By day 28, mortality rises with every step down in consciousness:
- Clear: 1.2% (23 of 1,974)
- ResponsiveToSound: 15.8% (3 of 19)
- ResponsiveToPain: 25.0% (1 of 4)
- Nonresponsive: 90.9% (10 of 11)

**Death comes early.** 67.6% of all 28-day deaths happen in week 1. Nonresponsive patients reach 54.5% mortality by day 7, 72.7% by day 14 and 81.8% by day 21. ResponsiveToSound and ResponsiveToPain patients who died did so within the first 5 days, so their lines are flat after week 1.

**Impaired consciousness is rare but lethal.** Only 1.7% of patients (34) arrive with impaired consciousness. Their 28-day mortality is 41.2% vs 1.2% for clear patients, a relative risk of about 35× (Fisher's exact p ≈ 4×10⁻¹⁷). They account for 37.8% of all 28-day deaths.

**Caveats:**
- The impaired groups are very small (4–19 patients), so their exact percentages are imprecise. The overall direction is still very clear.
- The consciousness label does not catch everyone. 23 patients labelled "Clear" have a GCS below 15, so the Clear line slightly understates risk for truly alert patients.

### How it helps analysis:

Shows that the danger from impaired consciousness is immediate: the first 7 days are the critical window, so escalation (ICU review, perfusion and oxygen checks) must happen on admission, not after days of observation.

Supports using consciousness level (or GCS) as a strong early-mortality predictor in the predictive models, ideally as an ordered feature, since risk climbs with each level.

Identifies Nonresponsive patients as a group where early goals-of-care discussions with families are appropriate, given that 9 in 10 die within 4 weeks.

Confirms the cleaned death timing (death_time_days) is complete and consistent with the death_28d flag, so time-to-event analysis is reliable for this outcome.
