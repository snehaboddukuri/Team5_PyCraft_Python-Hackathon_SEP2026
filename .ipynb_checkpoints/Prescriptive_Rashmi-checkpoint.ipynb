## 1 - Which medication combination should be prescribed to maximize 6-month survival?
### Reasoning

Heart-failure patients often receive multiple medications simultaneously. Instead of analyzing individual drugs alone, medication combinations were evaluated to identify treatment patterns associated with improved 6-month survival. Using survival outcomes helps determine which therapies may contribute to better long-term clinical results.

### Metrics, Biomarkers and Features Used
heart_pumping_efficiency
cci_score
heart_failure_severity
heart_damage_level
brain_natriuretic_peptide
creatinine
gcs
num_distinct_drugs

### Why These Were Used
heart-failure severity variables explain disease progression.
brain_natriuretic_peptide, Creatinine, and GCS represent major physiological risk factors.
cci_score captures overall comorbidity burden.

Together they provide both treatment and patient-risk context.

```python
# Step 1 - Get Medication column

med_cols = [col for col in df.columns if col.startswith('drug_')]

print(f"Total medications: {len(med_cols)}")
```

```python
# Step 2 - select most common medication

med_freq = df[med_cols].sum().sort_values(
    ascending=False
)

top_medications = med_freq.head(10).index.tolist()

print(top_medications)
```

```python
# Step 3 - create medication combination label

survival_df = df.copy()

survival_df['medication_combination'] = (
    survival_df[top_medications]
    .apply(
        lambda row: ', '.join(
            [
                col.replace('drug_', '')
                for col in top_medications
                if row[col] == 1
            ]
        ),
        axis=1
    )
)

survival_df['medication_combination'] = (
    survival_df['medication_combination']
    .replace('', 'No_Medication')
)
```

```python
# Step 4 - Calculate survival rate

med_survival = (
    survival_df.groupby(
        'medication_combination'
    )
    .agg(
        Patients=('patient_id', 'count'),
        Deaths=('death_6m', 'sum')
    )
)

med_survival = med_survival[
    med_survival['Patients'] >= 10
]

med_survival['Survival_Rate'] = (
    (
        med_survival['Patients']
        -
        med_survival['Deaths']
    )
    /
    med_survival['Patients']
    * 100
)

med_survival = (
    med_survival
    .sort_values(
        'Survival_Rate',
        ascending=False
    )
)

print(
    med_survival.head(20)
)
```

```python
# Top surviver medication combination chart

import seaborn as sns
import matplotlib.pyplot as plt

top10 = med_survival.head(10)

plt.figure(figsize=(12,8))

sns.barplot(
    data=top10,
    x='Survival_Rate',
    y=top10.index
)

plt.xlabel("6-Month Survival Rate (%)")
plt.ylabel("Medication Combination")

plt.title(
    "Top Medication Combinations Associated with Survival"
)

plt.show()
```

```python
# to determine which medications contribute most to survival.
from sklearn.ensemble import RandomForestClassifier

X = df[med_cols]

y = df['death_6m']

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42
)

rf.fit(X, y)

importance_df = pd.DataFrame({
    'Medication': med_cols,
    'Importance': rf.feature_importances_
})

importance_df = (
    importance_df
    .sort_values(
        'Importance',
        ascending=False
    )
)

print(
    importance_df.head(15)
)
```

### Key Insights

Expected high-survival combinations are likely to include:

Guideline-Based Heart Failure Therapy

Benazepril or Valsartan
Cardiovascular Protection

Patients receiving combinations aligned with standard heart-failure management pathways are generally expected to show better outcomes than patients requiring rescue therapies such as Milrinone, Dobutamine, or Isoprenaline, which often indicate severe disease.

## 2 - Which Laboratory Abnormalities Should Trigger Early Intervention?

### Reasoning
Laboratory abnormalities often appear before visible clinical deterioration. Identifying abnormal biomarkers early enables clinicians to intervene before a patient progresses to severe heart failure, readmission, or mortality. Since your merged dataset contains renal, cardiac, hematologic, inflammatory, electrolyte, and blood-gas parameters along with mortality outcomes (death_6m), these biomarkers can be used to define intervention thresholds.

### Metrics, Biomarkers, and Features Used
brain_natriuretic_peptide, high_sensitivity_troponin, myoglobin, creatine_kinase_isoenzyme,  creatinine, urea, glomerular_filtration_rate, cystatin, uric_acid, hemoglobin, WBC
platelet, neutrophil_ratio, lymphocyte_count, sodium, potassium, calcium, magnesium
chloride, lactate, pH, pO2, sO2, anion_gap.

### Why These Were Used

These biomarkers reflect the most important organ systems involved in heart failure:

BNP indicates cardiac stress.
Troponin indicates myocardial injury.
Creatinine / GFR indicate kidney dysfunction.
Hemoglobin reflects oxygen delivery capacity.
Lactate reflects tissue hypoperfusion.
Sodium often falls in severe heart failure.
Potassium abnormalities may trigger fatal arrhythmias.

These markers are clinically meaningful, explainable, and actionable.

###
### 

```python
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

lab_cols = [
    'brain_natriuretic_peptide',
    'high_sensitivity_troponin',
    'creatinine',
    'glomerular_filtration_rate',
    'urea',
    'hemoglobin',
    'sodium',
    'potassium',
    'lactate',
    'WBC',
    'platelet'
]

X = df[lab_cols].fillna(
    df[lab_cols].median()
)

y = df['death_6m']

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42
)

rf.fit(X, y)

importance_df = pd.DataFrame({
    'Biomarker': lab_cols,
    'Importance': rf.feature_importances_
})

importance_df = (
    importance_df
    .sort_values(
        'Importance',
        ascending=False
    )
)

print(importance_df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

top10 = importance_df.head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    data=top10,
    x='Importance',
    y='Biomarker'
)

plt.title(
    'Top Laboratory Markers Associated with Mortality'
)

plt.show()
```

### Key Insights
1.Patients with higher BNP levels tend to have a greater risk of mortality, indicating increased stress on the heart.

2.Patients with elevated creatinine and reduced kidney function (GFR) generally show poorer outcomes, highlighting the importance of kidney health in heart-failure management. 

3.Patients with abnormal sodium and potassium levels are more likely to experience complications and require closer monitoring.

4.Higher troponin levels are associated with increased cardiac injury and a higher probability of mortality.

5.Patients with low hemoglobin levels often demonstrate worse clinical outcomes due to reduced oxygen-carrying capacity.

### Prescriptive Actions to Be Taken

1.If BNP is significantly elevated, schedule immediate cardiac assessment and increase monitoring frequency.

2.If creatinine rises or GFR decreases, initiate kidney-protection protocols and review medication dosages.

3.If sodium or potassium levels become abnormal, trigger electrolyte correction and continuous cardiac monitoring.

4.If troponin levels increase, prioritize the patient for rapid cardiovascular evaluation and risk stratification.

5.If hemoglobin levels drop below normal, investigate anemia and consider interventions to improve oxygen delivery.

### Step 1: Identify High-Risk Groups by Readmission

```python
adherence_risk = (
    df.groupby('readmission_6m')
      .agg(
          Avg_Drugs=('num_distinct_drugs', 'mean'),
          Avg_CCI=('cci_score', 'mean'),
          Avg_Visits=('visit_count', 'mean')
      )
)

print(adherence_risk)
```

### Step 2: Medication Burden vs Readmission

```python
drug_burden = (
    df.groupby('num_distinct_drugs')
      ['readmission_6m']
      .mean()
      .mul(100)
      .reset_index()
)

drug_burden.columns = [
    'Number_of_Drugs',
    'Readmission_Rate'
]

print(drug_burden)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

sns.lineplot(
    data=drug_burden,
    x='Number_of_Drugs',
    y='Readmission_Rate',
    marker='o'
)

plt.title(
    'Readmission Rate by Medication Burden'
)

plt.ylabel('Readmission Rate (%)')

plt.show()
```

### Step 3: CCI Score vs Readmission

```python
cci_readmission = (
    df.groupby('cci_score')
      ['readmission_6m']
      .mean()
      .mul(100)
      .reset_index()
)

print(cci_readmission)
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=cci_readmission,
    x='cci_score',
    y='readmission_6m'
)

plt.title(
    'Readmission Rate by CCI Score'
)

plt.ylabel('Readmission Rate (%)')

plt.show()
```

### Step 4: Age Groups Needing Adherence Support

```python
age_adherence = (
    df.groupby('age_category')
      .agg(
          Readmission_Rate=('readmission_6m',
                            lambda x: x.mean()*100),
          Avg_Drugs=('num_distinct_drugs',
                     'mean')
      )
      .reset_index()
)

print(age_adherence)
```

### Step 5: Severe HF Patients

```python
severe_hf_analysis = (
    df.groupby('severe_hf')
      .agg(
          Readmission_Rate=('readmission_6m',
                            lambda x: x.mean()*100),
          Mortality_Rate=('death_6m',
                          lambda x: x.mean()*100)
      )
)

print(severe_hf_analysis)
```

### Final Adherence Flag

```python
df['Need_Adherence_Program'] = (
    (
        (df['num_distinct_drugs'] >= 8)
        |
        (df['cci_score'] >= 3)
        |
        (df['severe_hf'] == 1)
        |
        (df['visit_count'] >= 2)
    )
).astype(int)

df['Need_Adherence_Program'].value_counts()
```

### Metrics, Biomarkers and Features used

### Why These Were Used?

GCS → neurological condition.

```python
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

features = [
    'cci_score',
    'heart_pumping_efficiency',
    'heart_failure_severity',
    'heart_damage_level',
    'severe_hf',
    'gcs',
    'brain_natriuretic_peptide',
    'high_sensitivity_troponin',
    'creatinine',
    'glomerular_filtration_rate',
    'hemoglobin',
    'sodium',
    'potassium',
    'visit_count',
    'num_distinct_drugs'
]

X = df[features].fillna(df[features].median())
y = df['death_6m']

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42
)

rf.fit(X, y)

importance_df = pd.DataFrame({
    'Variable': features,
    'Importance': rf.feature_importances_
})

importance_df = importance_df.sort_values(
    'Importance',
    ascending=False
)

print(importance_df.head(15))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

top15 = importance_df.head(15)

plt.figure(figsize=(10,6))

sns.barplot(
    data=top15,
    x='Importance',
    y='Variable'
)

plt.title(
    'Top Predictors of 6-Month Mortality'
)

plt.show()
```

### Key Insights

Potassium is the strongest driver of mortality.

The presence of both glomerular_filtration_rate and creatinine among the top predictors suggests that renal dysfunction plays a major role in patient outcomes.

Electrolyte imbalance significantly impacts survival.

Potassium and sodium emerged as top predictors, indicating that electrolyte abnormalities are strongly associated with mortality risk.

Neurological condition matters.

Low hemoglobin appears to be an important predictor, suggesting reduced oxygen-carrying capacity may worsen prognosis.

Cardiac injury and comorbidity burden remain critical.

Elevated troponin and higher CCI scores indicate increased mortality risk due to cardiac damage and multiple chronic diseases.
