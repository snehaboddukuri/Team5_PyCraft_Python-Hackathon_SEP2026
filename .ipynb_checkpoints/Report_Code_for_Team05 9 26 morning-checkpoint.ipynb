### D1 · What is the age and sex profile of admitted heart-failure patients?

```python
# Self-contained: does not need the setup cell
# ---- Setup: libraries and paths (run the notebook from notebooks/descriptive/) ----
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats
import warnings; warnings.filterwarnings('ignore')
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'   # Team05 wide merged table (the only data source)
NAVY, RED, TEAL, AMBER = '#13294B', '#B3261E', '#1F7A74', '#B7791F'

# ---- Helper written out in full: event rate of a 0/1 outcome by group + chi-square test ----
def rate_by_group(data, group, outcome):
    d = data[[group, outcome]].dropna()
    table = d.groupby(group, observed=True)[outcome].agg(n='size', events='sum')
    table['rate_%'] = (table['events'] / table['n'] * 100).round(1)
    contingency = pd.crosstab(d[group], d[outcome])            # groups x (0/1) counts
    chi2, p, dof, _ = stats.chi2_contingency(contingency)      # H0: outcome rate is the same in every group
    return table, p

def fmt_p(p):
    return '<0.001' if p < 0.001 else f'{p:.3f}'

demo = pd.read_csv(WIDE, usecols=['patient_id', 'gender', 'weight', 'height', 'bmi', 'occupation', 'age_category'])
print(demo.shape)
demo[['patient_id', 'age_category', 'gender']].head()

band_order = ['21-29', '29-39', '39-49', '49-59', '59-69', '69-79', '79-89', '89-110']
demo['age_category'] = pd.Categorical(demo['age_category'], categories=band_order, ordered=True)
demo['age_lower'] = demo['age_category'].astype(str).str.split('-').str[0].astype(int)
demo[['age_category', 'age_lower']].drop_duplicates().sort_values('age_lower')

n = len(demo)
band_pct = demo['age_category'].value_counts(normalize=True).sort_index() * 100
female_pct = (demo['gender'] == 'Female').mean() * 100
over70 = (demo['age_lower'] >= 69).mean() * 100
over80 = (demo['age_lower'] >= 79).mean() * 100
fem_over80 = (demo.loc[demo['gender'] == 'Female', 'age_lower'] >= 79).mean() * 100
male_over80 = (demo.loc[demo['gender'] == 'Male', 'age_lower'] >= 79).mean() * 100
# Is the age distribution different between the sexes? (chi-square on the band x sex table)
chi2, p_age_sex, _, _ = stats.chi2_contingency(pd.crosstab(demo['age_category'], demo['gender']))
print(f'n = {n} | female {female_pct:.1f}% | >=70 y {over70:.1f}% | >=80 y {over80:.1f}% | women >=80 {fem_over80:.1f}% vs men {male_over80:.1f}% | p = {p_age_sex:.4f}')
pd.crosstab(demo['age_category'], demo['gender'], margins=True)

fig, ax = plt.subplots(figsize=(8, 3.5))
pd.crosstab(demo['age_category'], demo['gender']).plot.bar(ax=ax, color=[RED, NAVY])
ax.set_title('Admissions by age band and sex'); ax.set_ylabel('patients'); ax.set_xlabel('age band (years)'); plt.xticks(rotation=0); plt.tight_layout(); plt.show()

assert demo['patient_id'].is_unique
assert demo['age_category'].notna().all() and demo['gender'].isin(['Male', 'Female']).all()
print('checks passed: unique IDs, no missing age band or sex')
```

### D2 · What does the BMI distribution look like, and is under-nutrition a bigger issue than obesity?

```python
# Self-contained: does not need the setup cell
# ---- Setup: libraries and paths (run the notebook from notebooks/descriptive/) ----
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats
import warnings; warnings.filterwarnings('ignore')
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'   # Team05 wide merged table (the only data source)
NAVY, RED, TEAL, AMBER = '#13294B', '#B3261E', '#1F7A74', '#B7791F'

# ---- Helper written out in full: event rate of a 0/1 outcome by group + chi-square test ----
def rate_by_group(data, group, outcome):
    d = data[[group, outcome]].dropna()
    table = d.groupby(group, observed=True)[outcome].agg(n='size', events='sum')
    table['rate_%'] = (table['events'] / table['n'] * 100).round(1)
    contingency = pd.crosstab(d[group], d[outcome])            # groups x (0/1) counts
    chi2, p, dof, _ = stats.chi2_contingency(contingency)      # H0: outcome rate is the same in every group
    return table, p

def fmt_p(p):
    return '<0.001' if p < 0.001 else f'{p:.3f}'

demo = pd.read_csv(WIDE, usecols=['patient_id', 'gender', 'weight', 'height', 'bmi', 'occupation', 'age_category'])
hosp = pd.read_csv(WIDE, usecols=['patient_id', 'death_6m'])
df = demo.merge(hosp, on='patient_id', how='inner', validate='one_to_one')
print(df.shape, '| BMI missing:', df['bmi'].isna().sum())

bmi_labels = ['Underweight', 'Normal', 'Overweight', 'Obese']
df['bmi_category'] = pd.cut(df['bmi'], bins=[0, 18.5, 23, 27.5, np.inf], labels=bmi_labels, right=False)
b = df['bmi'].dropna()
share = df['bmi_category'].value_counts(normalize=True).reindex(bmi_labels) * 100
print(b.describe().round(2).to_string()); print(share.round(1).to_string())

t_bmi, p_bmi = rate_by_group(df, 'bmi_category', 'death_6m')
print('chi-square p =', round(p_bmi, 3)); t_bmi

fig, ax = plt.subplots(figsize=(8, 3.3)); ax.hist(b, bins=40, color=NAVY)
for x in [18.5, 23, 27.5]: ax.axvline(x, color='k', ls='--', lw=1)
ax.set_title('BMI distribution with Asian WHO cut-offs 18.5 / 23 / 27.5'); ax.set_xlabel('kg/m²'); ax.set_ylabel('patients'); plt.tight_layout(); plt.show()

assert (b.between(10, 60)).all(), 'implausible BMI left in data'
print('checks passed: all BMI values between 10 and 60')
```

### D3 · How severe is heart failure at admission (NYHA class and Killip grade)?

```python
# Self-contained: does not need the setup cell
# ---- Setup: libraries and paths (run the notebook from notebooks/descriptive/) ----
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats
import warnings; warnings.filterwarnings('ignore')
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'   # Team05 wide merged table (the only data source)
NAVY, RED, TEAL, AMBER = '#13294B', '#B3261E', '#1F7A74', '#B7791F'

# ---- Helper written out in full: event rate of a 0/1 outcome by group + chi-square test ----
def rate_by_group(data, group, outcome):
    d = data[[group, outcome]].dropna()
    table = d.groupby(group, observed=True)[outcome].agg(n='size', events='sum')
    table['rate_%'] = (table['events'] / table['n'] * 100).round(1)
    contingency = pd.crosstab(d[group], d[outcome])            # groups x (0/1) counts
    chi2, p, dof, _ = stats.chi2_contingency(contingency)      # H0: outcome rate is the same in every group
    return table, p

def fmt_p(p):
    return '<0.001' if p < 0.001 else f'{p:.3f}'

card = pd.read_csv(WIDE, usecols=['patient_id', 'heart_failure_severity', 'heart_damage_level'])
hosp = pd.read_csv(WIDE, usecols=['patient_id', 'death_28d', 'death_6m'])
df = card.merge(hosp, on='patient_id', validate='one_to_one'); print(df.shape); print(df.isna().sum().to_dict())

nyha = df['heart_failure_severity'].value_counts(normalize=True).sort_index() * 100
kil = df['heart_damage_level'].value_counts(normalize=True).sort_index() * 100
pd.DataFrame({'NYHA %': nyha.round(1)}).join(pd.DataFrame({'Killip %': kil.round(1)}), how='outer')

t_k, p_k = rate_by_group(df, 'heart_damage_level', 'death_28d')
t_ny, p_ny = rate_by_group(df, 'heart_failure_severity', 'death_6m')
print('28-day death by Killip, p =', fmt_p(p_k)); print(t_k); print('\n6-month death by NYHA, p =', fmt_p(p_ny)); print(t_ny)

fig, ax = plt.subplots(1, 2, figsize=(10, 3.3))
ax[0].bar(nyha.index.astype(str), nyha, color=NAVY); ax[0].set_title('NYHA class (% of patients)')
ax[1].bar(t_k.index.astype(str), t_k['rate_%'], color=RED); ax[1].set_title('28-day death by Killip grade (%)')
for i, v in enumerate(t_k['rate_%']): ax[1].text(i, v, f'{v}%', ha='center', va='bottom')
plt.tight_layout(); plt.show()

assert set(df['heart_damage_level'].unique()) <= {1, 2, 3, 4} and set(df['heart_failure_severity'].unique()) <= {2, 3, 4}
print('checks passed: grades within valid ranges')
```

```python
# Self-contained: does not need the setup cell
# ---- Setup: libraries and paths (run the notebook from notebooks/descriptive/) ----
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats
import warnings; warnings.filterwarnings('ignore')
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'   # Team05 wide merged table (the only data source)
NAVY, RED, TEAL, AMBER = '#13294B', '#B3261E', '#1F7A74', '#B7791F'

card = pd.read_csv(WIDE, usecols=['patient_id', 'heart_pumping_efficiency'])
demo = pd.read_csv(WIDE, usecols=['patient_id', 'gender'])
df = card.merge(demo, on='patient_id', validate='one_to_one')
print(df.shape, '| LVEF recorded:', df['heart_pumping_efficiency'].notna().sum(), f"({df['heart_pumping_efficiency'].notna().mean()*100:.1f}%)")

e = df[df['heart_pumping_efficiency'].notna()].copy()
e['hf_phenotype'] = pd.cut(e['heart_pumping_efficiency'], bins=[0, 40, 50, 100], labels=['HFrEF', 'HFmrEF', 'HFpEF'], right=False)
ph_share = e['hf_phenotype'].value_counts(normalize=True).reindex(['HFrEF', 'HFmrEF', 'HFpEF']) * 100
by_sex = pd.crosstab(e['gender'], e['hf_phenotype'], normalize='index') * 100
chi2, p_sex, _, _ = stats.chi2_contingency(pd.crosstab(e['gender'], e['hf_phenotype']))
print(e['heart_pumping_efficiency'].describe().round(1).to_string()); print(ph_share.round(1).to_string()); print('sex difference chi-square p =', round(p_sex, 5))
by_sex.round(1)

fig, ax = plt.subplots(figsize=(7, 3.3)); by_sex.plot.barh(stacked=True, ax=ax, color=[RED, AMBER, TEAL])
ax.set_xlabel('% of patients with an echo'); ax.set_title(f'HF phenotype by sex (n = {len(e)} with LVEF)'); ax.legend(loc='lower right', fontsize=8); plt.tight_layout(); plt.show()

assert e['heart_pumping_efficiency'].between(5, 90).all(); assert e['hf_phenotype'].notna().all()
print('checks passed: LVEF within 5-90% and every echo patient classified')
```

### D5 · How high is BNP (the heart-strain hormone), and does the level carry prognostic information?

```python
# Self-contained: does not need the setup cell
# ---- Setup: libraries and paths (run the notebook from notebooks/descriptive/) ----
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats
import warnings; warnings.filterwarnings('ignore')
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'   # Team05 wide merged table (the only data source)
NAVY, RED, TEAL, AMBER = '#13294B', '#B3261E', '#1F7A74', '#B7791F'

# ---- Helper written out in full: event rate of a 0/1 outcome by group + chi-square test ----
def rate_by_group(data, group, outcome):
    d = data[[group, outcome]].dropna()
    table = d.groupby(group, observed=True)[outcome].agg(n='size', events='sum')
    table['rate_%'] = (table['events'] / table['n'] * 100).round(1)
    contingency = pd.crosstab(d[group], d[outcome])            # groups x (0/1) counts
    chi2, p, dof, _ = stats.chi2_contingency(contingency)      # H0: outcome rate is the same in every group
    return table, p

def fmt_p(p):
    return '<0.001' if p < 0.001 else f'{p:.3f}'

labs = pd.read_csv(WIDE, usecols=['patient_id', 'brain_natriuretic_peptide'])
hosp = pd.read_csv(WIDE, usecols=['patient_id', 'death_6m'])
df = labs.merge(hosp, on='patient_id', validate='one_to_one'); print(df.shape, '| BNP missing:', df['brain_natriuretic_peptide'].isna().sum())

bnp = df['brain_natriuretic_peptide'].dropna()
above100 = (bnp > 100).mean() * 100; above1000 = (bnp >= 1000).mean() * 100
print(bnp.describe().round(0).to_string()); print(f'> 100 pg/mL: {above100:.1f}% | >= 1000 pg/mL: {above1000:.1f}%')

df['bnp_quartile'] = pd.qcut(df['brain_natriuretic_peptide'], 4, labels=['Q1', 'Q2', 'Q3', 'Q4'])
cuts = df.groupby('bnp_quartile', observed=True)['brain_natriuretic_peptide'].agg(['min', 'max']).round(0)
t_b, p_b = rate_by_group(df, 'bnp_quartile', 'death_6m')
print(cuts); print('chi-square p =', fmt_p(p_b)); t_b

fig, ax = plt.subplots(figsize=(6.5, 3.3)); ax.bar(t_b.index.astype(str), t_b['rate_%'], color=RED)
for i, v in enumerate(t_b['rate_%']): ax.text(i, v, f'{v}%', ha='center', va='bottom')
ax.set_title('6-month mortality by BNP quartile'); ax.set_ylabel('% died'); plt.tight_layout(); plt.show()

assert bnp.min() > 0 and bnp.max() <= 5000
print('checks passed: BNP positive and within the assay ceiling (5000 pg/mL)')
```

### D6 · How common are kidney impairment and anaemia — the "cardio-renal-anaemia" triad?

```python
# Self-contained: does not need the setup cell
# ---- Setup: libraries and paths (run the notebook from notebooks/descriptive/) ----
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats
import warnings; warnings.filterwarnings('ignore')
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'   # Team05 wide merged table (the only data source)
NAVY, RED, TEAL, AMBER = '#13294B', '#B3261E', '#1F7A74', '#B7791F'

# ---- Helper written out in full: event rate of a 0/1 outcome by group + chi-square test ----
def rate_by_group(data, group, outcome):
    d = data[[group, outcome]].dropna()
    table = d.groupby(group, observed=True)[outcome].agg(n='size', events='sum')
    table['rate_%'] = (table['events'] / table['n'] * 100).round(1)
    contingency = pd.crosstab(d[group], d[outcome])            # groups x (0/1) counts
    chi2, p, dof, _ = stats.chi2_contingency(contingency)      # H0: outcome rate is the same in every group
    return table, p

def fmt_p(p):
    return '<0.001' if p < 0.001 else f'{p:.3f}'

labs = pd.read_csv(WIDE, usecols=['patient_id', 'glomerular_filtration_rate', 'hemoglobin'])
demo = pd.read_csv(WIDE, usecols=['patient_id', 'gender'])
df = labs.merge(demo, on='patient_id', validate='one_to_one'); print(df.shape); print(df.isna().sum().to_dict())

stages = ['G5', 'G4', 'G3b', 'G3a', 'G2', 'G1']
df['ckd_stage'] = pd.cut(df['glomerular_filtration_rate'], bins=[0, 15, 30, 45, 60, 90, np.inf], labels=stages, right=False)
egfr = df['glomerular_filtration_rate'].dropna()
low_egfr = (egfr < 60).mean() * 100; severe = (egfr < 30).mean() * 100
stage_pct = df['ckd_stage'].value_counts(normalize=True).reindex(stages[::-1]) * 100
print(stage_pct.round(1).to_string()); print(f'eGFR < 60: {low_egfr:.1f}% | eGFR < 30: {severe:.1f}%')

threshold = np.where(df['gender'] == 'Male', 130, 120)
df['anaemia'] = np.where(df['hemoglobin'].isna(), np.nan, (df['hemoglobin'] < threshold).astype(float))
anaemia_pct = df['anaemia'].mean() * 100
both = df.dropna(subset=['anaemia', 'glomerular_filtration_rate'])
both_pct = ((both['anaemia'] == 1) & (both['glomerular_filtration_rate'] < 60)).mean() * 100
# Are the two conditions associated? (chi-square on the 2x2 table)
chi2, p_assoc, _, _ = stats.chi2_contingency(pd.crosstab(both['anaemia'], both['glomerular_filtration_rate'] < 60))
anaemia_by_kidney = both.groupby(both['glomerular_filtration_rate'] < 60)['anaemia'].mean() * 100
print(f'anaemia {anaemia_pct:.1f}% | both {both_pct:.1f}% | anaemia with eGFR<60 {anaemia_by_kidney[True]:.1f}% vs {anaemia_by_kidney[False]:.1f}% | p = {p_assoc:.2g}')

fig, ax = plt.subplots(figsize=(7, 3.3)); ax.bar(stage_pct.index.astype(str), stage_pct, color=[TEAL, TEAL, AMBER, AMBER, RED, RED])
ax.set_title('Kidney function (KDIGO stage), % of patients'); ax.set_ylabel('%'); plt.tight_layout(); plt.show()

assert egfr.between(0, 300).all(); assert df['hemoglobin'].dropna().between(25, 220).all()
print('checks passed: eGFR 0-300 (Team05 does not cap eGFR) and Hb 25-220 g/L (Team05 PLAUSIBLE_LIMITS)')
```

### D7 · What happens to patients during and after the admission (length of stay, death, readmission)?

```python
# Self-contained: does not need the setup cell
# ---- Setup: libraries and paths (run the notebook from notebooks/descriptive/) ----
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats
import warnings; warnings.filterwarnings('ignore')
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'   # Team05 wide merged table (the only data source)
NAVY, RED, TEAL, AMBER = '#13294B', '#B3261E', '#1F7A74', '#B7791F'

hosp = pd.read_csv(WIDE, usecols=['patient_id', 'discharge_destination', 'admission_ward', 'admission_type', 'discharge_department', 'visit_count', 'respiratory_support', 'oxygen_therapy', 'discharge_day', 'admission_date', 'hospitalization_outcome', 'death_28d', 'readmission_28d', 'death_3m', 'readmission_3m', 'death_6m', 'readmission_6m', 'death_time_days', 'readmission_time_days', 'emergency_return_6m', 'emergency_return_time_days'])
print(hosp.shape)

los = hosp['discharge_day']
long_stay = (los >= 14).mean() * 100      # 'prolonged stay' = 14 days or more
print(los.describe().round(1).to_string()); print(f'>= 14 days: {long_stay:.1f}%')

windows = ['28 days', '3 months', '6 months']
win = pd.DataFrame({
    'Death %':       [hosp[c].mean() * 100 for c in ['death_28d', 'death_3m', 'death_6m']],
    'Readmission %': [hosp[c].mean() * 100 for c in ['readmission_28d', 'readmission_3m', 'readmission_6m']]},
    index=windows).round(1)
ratio = win.loc['6 months', 'Readmission %'] / win.loc['6 months', 'Death %']
print(f'readmissions per death at 6 months: {ratio:.1f}'); win

fig, ax = plt.subplots(1, 2, figsize=(10, 3.3))
ax[0].hist(los.clip(upper=40), bins=40, color=NAVY); ax[0].set_title('Length of stay (days, capped at 40 for display)')
win.plot(ax=ax[1], marker='o', color=[RED, NAVY]); ax[1].set_title('Cumulative outcome (%)'); ax[1].set_ylabel('%')
plt.tight_layout(); plt.show()

assert (win['Death %'].is_monotonic_increasing) and (win['Readmission %'].is_monotonic_increasing)
print('checks passed: cumulative rates never decrease over time (nested flags are consistent)')
```

### D9 · How many patients arrive with impaired consciousness, and how dangerous is it?

```python
# Self-contained: does not need the setup cell
# ---- Setup: libraries and paths (run the notebook from notebooks/descriptive/) ----
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats
import warnings; warnings.filterwarnings('ignore')
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'   # Team05 wide merged table (the only data source)
NAVY, RED, TEAL, AMBER = '#13294B', '#B3261E', '#1F7A74', '#B7791F'

# ---- Helper written out in full: event rate of a 0/1 outcome by group + chi-square test ----
def rate_by_group(data, group, outcome):
    d = data[[group, outcome]].dropna()
    table = d.groupby(group, observed=True)[outcome].agg(n='size', events='sum')
    table['rate_%'] = (table['events'] / table['n'] * 100).round(1)
    contingency = pd.crosstab(d[group], d[outcome])            # groups x (0/1) counts
    chi2, p, dof, _ = stats.chi2_contingency(contingency)      # H0: outcome rate is the same in every group
    return table, p

def fmt_p(p):
    return '<0.001' if p < 0.001 else f'{p:.3f}'

resp = pd.read_csv(WIDE, usecols=['patient_id', 'gcs'])
hosp = pd.read_csv(WIDE, usecols=['patient_id', 'death_28d'])
df = resp.merge(hosp, on='patient_id', validate='one_to_one'); print(df.shape)

df['impaired'] = (df['gcs'] < 15).astype(int)
df['gcs_band'] = pd.cut(df['gcs'], bins=[0, 8, 12, 14, 15], labels=['Severe', 'Moderate', 'Mild', 'Normal'])
n_imp = df['impaired'].sum(); pct_imp = df['impaired'].mean() * 100; n_severe = (df['gcs'] <= 8).sum()
print(f'impaired: {n_imp} ({pct_imp:.1f}%) | severe (<=8): {n_severe}')
df['gcs_band'].value_counts().reindex(['Normal', 'Mild', 'Moderate', 'Severe']).to_frame('patients')

t_g, p_g = rate_by_group(df, 'impaired', 'death_28d')
rel_risk = t_g.loc[1, 'rate_%'] / t_g.loc[0, 'rate_%']
share_deaths = df.loc[df['impaired'] == 1, 'death_28d'].sum() / df['death_28d'].sum() * 100
# Fisher exact test as a check, because the impaired group is small
_, p_fisher = stats.fisher_exact(pd.crosstab(df['impaired'], df['death_28d']))
print(f'chi-square p = {fmt_p(p_g)} | Fisher p = {fmt_p(p_fisher)} | relative risk {rel_risk:.1f} | share of 28-day deaths {share_deaths:.0f}%'); t_g

fig, ax = plt.subplots(figsize=(6, 3.3)); ax.bar(['Alert (GCS 15)', 'Impaired (GCS < 15)'], t_g['rate_%'], color=[TEAL, RED])
for i, v in enumerate(t_g['rate_%']): ax.text(i, v, f'{v}%', ha='center', va='bottom')
ax.set_title('28-day mortality by consciousness at admission'); ax.set_ylabel('% died'); plt.tight_layout(); plt.show()

assert df['gcs'].between(3, 15).all()
print('checks passed: GCS within 3-15')
```

## Setup cell: run once before P1–P30 and H1–H50

```python
# ============================================================================
# SETUP CELL — run this ONCE in the Team05 notebook before the question code below.
# Reads Team05_PyCraft_Cleaned_data/wide_merged_table.csv, prepares the analysis
# table `df` in memory (nothing is written to disk) and defines the helper functions.
# Needs: pandas, numpy, matplotlib, scipy, scikit-learn  (pip install scipy scikit-learn)
# ============================================================================
import io, json, os, warnings; warnings.filterwarnings('ignore')
import pandas as pd, numpy as np, matplotlib.pyplot as plt

# ---- 1. Helper functions (statistics, models, charts) ----
"""Shared statistics & modelling helpers for the HF analytics notebooks.
Only pandas / numpy / scipy / scikit-learn / matplotlib are required."""
import numpy as np, pandas as pd
from scipy import stats
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression, LinearRegression, Ridge
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier, RandomForestRegressor
from sklearn.model_selection import StratifiedKFold, KFold, cross_val_predict, permutation_test_score
from sklearn.metrics import roc_auc_score, average_precision_score, balanced_accuracy_score, recall_score, confusion_matrix, r2_score, mean_absolute_error

PALETTE = {'main': '#8E2C3B', 'blue': '#3B5B8E', 'teal': '#2A7F7A', 'grey': '#7F7F7F', 'gold': '#C9A227'}

# ---------------------------------------------------------------- descriptive / association
def rate_table(df, group, outcome, min_n=1):
    """Event rate of a binary outcome by group, with chi-square p-value."""
    d = df[[group, outcome]].dropna()
    t = d.groupby(group, observed=True)[outcome].agg(n='size', events='sum', rate='mean')
    t['rate_%'] = (t['rate'] * 100).round(1); t = t.drop(columns='rate')
    t = t[t['n'] >= min_n]
    ct = pd.crosstab(d[group], d[outcome])
    p = stats.chi2_contingency(ct)[1] if ct.shape[0] > 1 and ct.shape[1] > 1 else np.nan
    return t, p

def compare_groups(df, var, outcome):
    """Median (IQR) of a continuous var by binary outcome + Mann-Whitney U p-value."""
    a = df.loc[df[outcome] == 1, var].dropna(); b = df.loc[df[outcome] == 0, var].dropna()
    p = stats.mannwhitneyu(a, b).pvalue if len(a) > 2 and len(b) > 2 else np.nan
    f = lambda s: f'{s.median():.2f} ({s.quantile(.25):.2f}-{s.quantile(.75):.2f})'
    return {'marker': var, 'outcome=1 median (IQR)': f(a), 'outcome=0 median (IQR)': f(b), 'n1': len(a), 'n0': len(b), 'p_value': p}

def screen_markers(df, markers, outcome):
    rows = [compare_groups(df, m, outcome) for m in markers]
    out = pd.DataFrame(rows).sort_values('p_value'); out['p_value'] = out['p_value'].map(lambda p: float(f'{p:.3g}'))
    return out.reset_index(drop=True)

def spearman(df, x, y):
    d = df[[x, y]].dropna(); r, p = stats.spearmanr(d[x], d[y]); return r, p, len(d)

def logit(df, outcome, predictors, standardise=False):
    """Unpenalised logistic regression with odds ratios, 95% CI and Wald p-values (statsmodels-style)."""
    d = df[[outcome] + predictors].dropna().astype(float)
    X = d[predictors].values; y = d[outcome].values
    if standardise: X = (X - X.mean(0)) / X.std(0)
    X1 = np.column_stack([np.ones(len(X)), X]); beta = np.zeros(X1.shape[1])
    for _ in range(100):  # Newton-Raphson
        p = 1 / (1 + np.exp(-X1 @ beta)); W = p * (1 - p)
        H = X1.T @ (X1 * W[:, None]) + 1e-8 * np.eye(len(beta)); g = X1.T @ (y - p)
        step = np.linalg.solve(H, g); beta += step
        if np.abs(step).max() < 1e-8: break
    se = np.sqrt(np.diag(np.linalg.inv(H))); z = beta / se; pv = 2 * stats.norm.sf(np.abs(z))
    res = pd.DataFrame({'OR': np.exp(beta), 'CI_low': np.exp(beta - 1.96 * se), 'CI_high': np.exp(beta + 1.96 * se), 'p_value': pv},
                       index=['intercept'] + predictors).iloc[1:]
    res.attrs['n'] = len(d); res.attrs['events'] = int(y.sum())
    return res.round(4)

# ---------------------------------------------------------------- modelling
def preprocessor(num_cols, cat_cols=()):
    tr = [('num', Pipeline([('imp', SimpleImputer(strategy='median', add_indicator=True)), ('sc', StandardScaler())]), list(num_cols))]
    if cat_cols:
        tr.append(('cat', Pipeline([('imp', SimpleImputer(strategy='most_frequent')), ('oh', OneHotEncoder(handle_unknown='ignore'))]), list(cat_cols)))
    return ColumnTransformer(tr)

def make_model(kind, num_cols, cat_cols=(), seed=42):
    if kind == 'logreg':
        est = LogisticRegression(max_iter=2000, class_weight='balanced', C=0.5)
    elif kind == 'rf':
        est = RandomForestClassifier(n_estimators=300, min_samples_leaf=10, class_weight='balanced_subsample', random_state=seed, n_jobs=-1)
    elif kind == 'hgb':
        est = HistGradientBoostingClassifier(max_depth=3, learning_rate=0.05, max_iter=200, class_weight='balanced', random_state=seed)
    else: raise ValueError(kind)
    return Pipeline([('prep', preprocessor(num_cols, cat_cols)), ('model', est)])

def evaluate_classifier(df, outcome, num_cols, cat_cols=(), kind='logreg', n_perm=0, seed=42, folds=5):
    """Stratified k-fold CV. Returns dict of AUC, PR-AUC, balanced accuracy, sensitivity, specificity (+ permutation p)."""
    d = df.dropna(subset=[outcome]); X = d[list(num_cols) + list(cat_cols)]; y = d[outcome].astype(int).values
    cv = StratifiedKFold(folds, shuffle=True, random_state=seed); mdl = make_model(kind, num_cols, cat_cols, seed)
    prob = cross_val_predict(mdl, X, y, cv=cv, method='predict_proba')[:, 1]
    thr = np.quantile(prob, 1 - y.mean()) if kind != 'logreg' else 0.5  # balanced models: 0.5; others: prevalence-matched
    pred = (prob >= thr).astype(int); tn, fp, fn, tp = confusion_matrix(y, pred).ravel()
    out = {'outcome': outcome, 'model': kind, 'n': len(y), 'events': int(y.sum()), 'prevalence_%': float(round(y.mean() * 100, 1)),
           'AUC': round(roc_auc_score(y, prob), 3), 'PR_AUC': round(average_precision_score(y, prob), 3),
           'baseline_PR_AUC': float(round(y.mean(), 3)), 'balanced_acc': round(balanced_accuracy_score(y, pred), 3),
           'sensitivity': float(round(tp / (tp + fn), 3)), 'specificity': float(round(tn / (tn + fp), 3))}
    lo, hi = auc_ci(prob, y); out['AUC_95CI'] = f'{lo:.3f}-{hi:.3f}'
    out['p_vs_chance'] = float(f"{stats.mannwhitneyu(prob[y == 1], prob[y == 0], alternative='greater').pvalue:.3g}")
    if n_perm:
        _, perm, p = permutation_test_score(mdl, X, y, cv=cv, scoring='roc_auc', n_permutations=n_perm, random_state=seed, n_jobs=-1)
        out['perm_mean_AUC'] = round(perm.mean(), 3); out['perm_p'] = round(p, 4)
    out['_prob'] = prob; out['_y'] = y
    return out

def top_features(df, outcome, num_cols, cat_cols=(), k=8):
    """Standardised logistic coefficients (fitted on all data) → direction + size of each marker."""
    d = df.dropna(subset=[outcome]); mdl = make_model('logreg', num_cols, cat_cols).fit(d[list(num_cols) + list(cat_cols)], d[outcome].astype(int))
    names = mdl.named_steps['prep'].get_feature_names_out(); coef = mdl.named_steps['model'].coef_[0]
    s = pd.Series(coef, index=[n.split('__', 1)[1] for n in names])
    s = s[~s.index.str.startswith('missingindicator')]
    return s.reindex(s.abs().sort_values(ascending=False).index).head(k).round(3)

def risk_deciles(prob, y, q=5):
    t = pd.DataFrame({'p': prob, 'y': y}); t['group'] = pd.qcut(t['p'].rank(method='first'), q, labels=[f'Q{i}' for i in range(1, q + 1)])
    return t.groupby('group', observed=True)['y'].agg(n='size', observed_rate='mean').assign(observed_rate=lambda x: (x.observed_rate * 100).round(1))

def evaluate_regressor(df, outcome, num_cols, cat_cols=(), seed=42):
    d = df.dropna(subset=[outcome]); X = d[list(num_cols) + list(cat_cols)]; y = d[outcome].values
    mdl = Pipeline([('prep', preprocessor(num_cols, cat_cols)), ('model', RandomForestRegressor(n_estimators=300, min_samples_leaf=10, random_state=seed, n_jobs=-1))])
    pred = cross_val_predict(mdl, X, y, cv=KFold(5, shuffle=True, random_state=seed))
    base = np.full_like(y, y.mean(), dtype=float)
    return {'outcome': outcome, 'n': len(y), 'R2': round(r2_score(y, pred), 3), 'MAE': round(mean_absolute_error(y, pred), 3),
            'MAE_baseline_mean': round(mean_absolute_error(y, base), 3), '_pred': pred, '_y': y}

def fmt_p(p):
    return '<0.001' if p < 0.001 else f'{p:.3f}'

def auc_ci(prob, y, n_boot=500, seed=0):
    """Bootstrap 95% CI of the out-of-fold AUC."""
    rng = np.random.default_rng(seed); aucs = []; idx = np.arange(len(y))
    for _ in range(n_boot):
        b = rng.choice(idx, len(idx), replace=True)
        if y[b].min() == y[b].max(): continue
        aucs.append(roc_auc_score(y[b], prob[b]))
    return np.percentile(aucs, 2.5), np.percentile(aucs, 97.5)

def hypothesis_verdict(res, min_auc=0.60):
    """Plain-language verdict. H0 = markers carry no predictive information (AUC = 0.5)."""
    lo = float(res['AUC_95CI'].split('-')[0])
    if res['p_vs_chance'] < 0.05 and lo > 0.5:
        strength = 'strong' if res['AUC'] >= 0.80 else 'moderate' if res['AUC'] >= 0.70 else 'weak' if res['AUC'] >= min_auc else 'very weak'
        return f"**Reject H0** (p = {fmt_p(res['p_vs_chance'])}; AUC 95% CI {res['AUC_95CI']} excludes 0.5). Discrimination is **{strength}** (AUC {res['AUC']})."
    return f"**Fail to reject H0** (p = {fmt_p(res['p_vs_chance'])}; AUC 95% CI {res['AUC_95CI']}). These markers alone do not predict the outcome better than chance."

ORDERS = {'agecat': ['21-29','29-39','39-49','49-59','59-69','69-79','79-89','89-110'], 'age_group': ['<70','70-79','>=80'],
          'bmi_cat': ['Underweight','Normal','Overweight','Obese'], 'hf_phenotype': ['HFrEF','HFmrEF','HFpEF','Unknown'],
          'ckd_stage': ['G1','G2','G3a','G3b','G4','G5'], 'bnp_quartile': ['Q1','Q2','Q3','Q4'], 'cci_group': ['0-1','2','3','4+'],
          'gcs_severity': ['Normal','Mild','Moderate','Severe'], 'anemia_severity': ['Mild/None (≥110)','Moderate (80-109)','Severe (<80)'],
          'season': ['Spring','Summer','Autumn','Winter']}

def load_combined(path='../data/clean/hf_combined_clean.csv'):
    """Load the merged clean dataset and restore ordered categories lost in CSV."""
    df = pd.read_csv(path, parse_dates=['admission_date'])
    for c, o in ORDERS.items():
        if c in df: df[c] = pd.Categorical(df[c], categories=o, ordered=True)
    return df

def savefig(fig, name):
    import os; os.makedirs('../figures', exist_ok=True); fig.savefig(f'../figures/{name}.png', dpi=130, bbox_inches='tight')

def bar_rates(t, title, name=None, ylabel='% with outcome', color=None, ax=None, annotate=True):
    """Bar chart of the rate_% column from rate_table, annotated with n."""
    own = ax is None
    if own: fig, ax = plt_subplots(figsize=(6.5, 3.2))
    x = [str(i) for i in t.index]
    bars = ax.bar(x, t['rate_%'], color=color or PALETTE['main'])
    if annotate:
        for b, (r, n) in zip(bars, zip(t['rate_%'], t['n'])):
            ax.text(b.get_x() + b.get_width() / 2, b.get_height(), f'{r}%\nn={n}', ha='center', va='bottom', fontsize=8)
    ax.set_title(title); ax.set_ylabel(ylabel); ax.set_ylim(0, max(t['rate_%']) * 1.3 + 0.5)
    if own:
        import matplotlib.pyplot as plt; plt.tight_layout()
        if name: savefig(ax.figure, name)
    return ax

def plt_subplots(**kw):
    import matplotlib.pyplot as plt; return plt.subplots(**kw)


def savefig(fig, name):
    """Charts are shown in the notebook; nothing is saved to disk."""
    pass

# ---- 2. Load the wide table and derive the analysis fields ----
import pandas as pd, numpy as np, json, os
WIDE = 'Team05_PyCraft_Cleaned_data/wide_merged_table.csv'
wide = pd.read_csv(WIDE)
print('wide table:', wide.shape)
# Columns of each Team05 cleaned table (as merged into the wide table)
TABLE_COLS = {"demo": ["patient_id", "gender", "weight", "height", "bmi", "occupation", "age_category"], "cardiac": ["patient_id", "heart_failure_severity", "heart_damage_level", "heart_attack_history", "congestive_heart_failure", "peripheral_vascular_disease", "heart_failure_type", "heart_pumping_efficiency", "left_ventricle_size", "mitral_valve_ems", "mitral_valve_ams", "ea_ratio", "pulmonary_velocity", "pulmonary_pressure", "severe_hf"], "hosp": ["patient_id", "discharge_destination", "admission_ward", "admission_type", "discharge_department", "visit_count", "respiratory_support", "oxygen_therapy", "discharge_day", "admission_date", "hospitalization_outcome", "death_28d", "readmission_28d", "death_3m", "readmission_3m", "death_6m", "readmission_6m", "death_time_days", "readmission_time_days", "emergency_return_6m", "emergency_return_time_days"], "labs": ["patient_id", "temp", "pulse", "resp", "sbp", "dbp", "MAP", "FiO2", "creatinine", "urea", "uric_acid", "glomerular_filtration_rate", "cystatin", "WBC", "monocyte_ratio", "monocyte_count", "RBC", "coefficient_of_variation_RBC_DistributionWidth", "Standard_Deviation_RBC_DistributionWidth", "MCV", "HCT", "lymphocyte_count", "MCH", "MCHC", "MPV", "basophil_ratio", "basophil_count", "eosinophil_ratio", "eosinophil_count", "hemoglobin", "platelet", "platelet_distribution_width", "platelet_hematocrit", "neutrophil_ratio", "neutrophil_count", "d_dimer", "international_normalized_ratio", "activated_partial_thromboplastin_time", "thrombin_time", "prothrombin_activity", "prothrombin_time_ratio", "fibrinogen", "high_sensitivity_troponin", "myoglobin", "CO2_capacity", "calcium", "potassium", "chloride", "sodium", "phosphorus", "magnesium", "creatine_kinase_isoenzyme_to_creatine_kinase", "hydroxybutyrate_dehydrogenase_to_lactate_dehydrogenase", "hydroxybutyrate_dehydrogenase", "glutamic_oxaloacetic_transaminase", "creatine_kinase", "creatine_kinase_isoenzyme", "lactate_dehydrogenase", "brain_natriuretic_peptide", "high_sensitivity_protein", "nucleotidase", "fucosidase", "albumin", "white_globulin_ratio", "glutamyltranspeptidase", "glutamic_pyruvic_transaminase", "glutamic_oxaliplatin", "indirect_bilirubin", "ALP", "globulin", "direct_bilirubin", "total_bilirubin", "total_bile_acid", "total_protein", "erythrocyte_sedimentation_rate", "cholesterol", "LDL_C", "triglyceride", "HDL_C", "homocysteine", "apolipoprotein_a", "apolipoprotein_b", "lipoprotein", "pH", "standard_residual_base", "standard_bicarbonate", "pCO2", "tCO2", "methemoglobin", "HCT_BG", "HHb", "K_ion", "Cl_ion", "Na_ion", "glucose_blood_gas", "lactate", "measured_residual_base", "HCO3_act", "CO_Hb", "temp_BG", "sO2", "pO2", "O2Hb", "anion_gap", "free_calcium", "total_hemoglobin"], "hist": ["patient_id", "cerebrovascular_disease", "dementia", "chronic_obstructive_pulmonary_disease", "connective_tissue_disease", "peptic_ulcer_disease", "diabetes", "moderate_to_severe_chronic_kidney_disease", "hemiplegia", "leukemia", "malignant_lymphoma", "solid_tumor", "liver_disease", "aids", "cci_score", "type2_respiratory_failure", "acute_renal_failure"], "resp": ["patient_id", "eye_opening", "verbal_response", "movement", "consciousness", "gcs"]}
t05 = {k: wide[cols].copy() for k, cols in TABLE_COLS.items()}
# Prescriptions: one row per patient x drug, rebuilt from the 0/1 drug_ flags
drug_cols = [c for c in wide.columns if c.startswith('drug_')]
rx_long = wide.melt(id_vars='patient_id', value_vars=drug_cols, var_name='drug_col', value_name='given')
rx_long = rx_long[rx_long['given'] == 1]
t05['rx'] = pd.DataFrame({'patient_id': rx_long['patient_id'].values,
                          'drug_name': rx_long['drug_col'].str.replace('drug_', '', n=1).str.replace('_', ' ').values})
for k, d in t05.items(): print(f'{k:8s} {d.shape}')
ids = set(wide['patient_id'])
assert wide['patient_id'].is_unique, 'duplicate patients in the wide table'
print('One row per patient, 2,008 unique IDs:', len(ids) == 2008)

MAP = {
 'all': {'patient_id': 'inpatient_number'},
 'demo': {'age_category': 'agecat'},
 'cardiac': {'heart_failure_severity': 'nyha_cardiac_function_classification', 'heart_damage_level': 'killip_grade', 'heart_attack_history': 'myocardial_infarction',
             'heart_failure_type': 'type_of_heart_failure', 'ea_ratio': 'ea', 'pulmonary_velocity': 'tricuspid_valve_return_velocity',
             'pulmonary_pressure': 'tricuspid_valve_return_pressure', 'heart_pumping_efficiency': 'lvef', 'left_ventricle_size': 'left_ventricular_end_diastolic_diameter_lv'},
 'hosp': {'admission_type': 'admission_way', 'discharge_destination': 'destinationdischarge', 'discharge_day': 'dischargeday', 'visit_count': 'visit_times',
          'oxygen_therapy': 'oxygen_inhalation', 'hospitalization_outcome': 'outcome_during_hospitalization', 'death_28d': 'death_within_28_days',
          'readmission_28d': 're_admission_within_28_days', 'death_3m': 'death_within_3_months', 'readmission_3m': 're_admission_within_3_months',
          'death_6m': 'death_within_6_months', 'readmission_6m': 're_admission_within_6_months', 'emergency_return_6m': 'return_to_emergency_department_within_6_months',
          'emergency_return_time_days': 'time_to_emergency_department_within_6_months', 'death_time_days': 'time_of_death__days_from_admission',
          'readmission_time_days': 'readmission_time_days_from_admission'},
 'labs': {'temp': 'body_temperature', 'resp': 'respiration', 'sbp': 'systolic_blood_pressure', 'dbp': 'diastolic_blood_pressure', 'MAP': 'map_value', 'FiO2': 'fio2',
          'creatinine': 'creatinine_enzymatic_method', 'WBC': 'white_blood_cell', 'RBC': 'red_blood_cell',
          'coefficient_of_variation_RBC_DistributionWidth': 'coefficient_of_variation_of_red_blood_cell_distribution_width',
          'Standard_Deviation_RBC_DistributionWidth': 'standard_deviation_of_red_blood_cell_distribution_width', 'MCV': 'mean_corpuscular_volume', 'HCT': 'hematocrit',
          'MCH': 'mean_hemoglobin_volume', 'MCHC': 'mean_hemoglobin_concentration', 'MPV': 'mean_platelet_volume', 'CO2_capacity': 'carbon_dioxide_binding_capacity',
          'phosphorus': 'inorganic_phosphorus', 'magnesium': 'serum_magnesium', 'ALP': 'alkaline_phosphatase', 'LDL_C': 'low_density_lipoprotein_cholesterol',
          'HDL_C': 'high_density_lipoprotein_cholesterol', 'pH': 'ph', 'pCO2': 'partial_pressure_of_carbon_dioxide', 'tCO2': 'total_carbon_dioxide',
          'HCT_BG': 'hematocrit_blood_gas', 'HHb': 'reduced_hemoglobin', 'K_ion': 'potassium_ion', 'Cl_ion': 'chloride_ion', 'Na_ion': 'sodium_ion',
          'HCO3_act': 'measured_bicarbonate', 'CO_Hb': 'carboxyhemoglobin', 'temp_BG': 'body_temperature_blood_gas', 'sO2': 'oxygen_saturation',
          'pO2': 'partial_oxygen_pressure', 'O2Hb': 'oxyhemoglobin'},
 'hist': {'type2_respiratory_failure': 'type_ii_respiratory_failure'},
 'resp': {}, 'rx': {},
}
t = {k: d.rename(columns={**MAP['all'], **MAP[k]}) for k, d in t05.items()}
mapping_table = pd.DataFrame([(k, a, b) for k in MAP if k != 'all' for a, b in MAP[k].items()], columns=['table', 'Team05 name', 'analysis name'])
print('Columns renamed:', len(mapping_table) + 1, '(incl. patient_id -> inpatient_number)')

demo = t['demo'].copy()
order = ['21-29','29-39','39-49','49-59','59-69','69-79','79-89','89-110']
mid = {'21-29':25,'29-39':34,'39-49':44,'49-59':54,'59-69':64,'69-79':74,'79-89':84,'89-110':95}
demo['age_mid'] = demo['agecat'].map(mid).astype(float)                       # numeric midpoint of the 10-year band
demo['age_group'] = pd.cut(demo['age_mid'], [0, 69, 79, 200], labels=['<70','70-79','>=80'])
demo['bmi_cat'] = pd.cut(demo['bmi'], [0, 18.5, 23, 27.5, 100], right=False, labels=['Underweight','Normal','Overweight','Obese'])   # Asian WHO cut-offs
# Team05 title-cased occupation ('Urbanresident'); restore the data-dictionary spelling
demo['occupation'] = demo['occupation'].str.lower().map({'urbanresident':'UrbanResident','farmer':'Farmer','worker':'Worker','officer':'Officer','others':'Others'}).fillna('Unknown')
print(demo['age_group'].value_counts().sort_index().to_dict(), demo['occupation'].value_counts().to_dict())

cc = t['cardiac'].drop(columns=['severe_hf'])          # Team05 flag = NYHA>=3 & Killip>=3; analysis uses the two scales directly
cc['echo_available'] = cc['lvef'].notna().astype(int)
cc['hf_phenotype'] = pd.cut(cc['lvef'], [0, 40, 50, 100], right=False, labels=['HFrEF','HFmrEF','HFpEF']).astype(object).fillna('Unknown')
cc['nyha_iv'] = (cc['nyha_cardiac_function_classification'] == 4).astype(int)
cc['killip_high'] = (cc['killip_grade'] >= 3).astype(int)
cc['pulm_htn'] = np.where(cc['tricuspid_valve_return_pressure'].isna(), np.nan, (cc['tricuspid_valve_return_pressure'] > 35).astype(float))
cc['restrictive_filling'] = np.where(cc['ea'].isna(), np.nan, (cc['ea'] > 2).astype(float))
print(cc['hf_phenotype'].value_counts().to_dict())

hd = t['hosp'].copy()
hd['admission_date'] = pd.to_datetime(hd['admission_date'])
hd['ventilated'] = hd['respiratory_support'].notna().astype(int)      # Team05 keeps blank = no ventilation (values: NIMV / IMV)
hd['adm_year'] = hd['admission_date'].dt.year; hd['adm_month'] = hd['admission_date'].dt.month
hd['season'] = hd['adm_month'].map({12:'Winter',1:'Winter',2:'Winter',3:'Spring',4:'Spring',5:'Spring',6:'Summer',7:'Summer',8:'Summer',9:'Autumn',10:'Autumn',11:'Autumn'})
hd['died_in_hospital'] = (hd['outcome_during_hospitalization'] == 'Dead').astype(int)
hd['outcome_conflict'] = (((hd['outcome_during_hospitalization']=='Alive') & (hd['destinationdischarge']=='Died')) |
                          ((hd['outcome_during_hospitalization']=='Dead') & (hd['destinationdischarge']!='Died'))).astype(int)
hd['log_los'] = np.log1p(hd['dischargeday'])
hd['prolonged_los'] = (hd['dischargeday'] >= 14).astype(int)
hd['death_or_readm_6m'] = ((hd['death_within_6_months']==1) | (hd['re_admission_within_6_months']==1)).astype(int)
hd['emergency_admission'] = (hd['admission_way'] == 'Emergency').astype(int)
hd['icu_admission'] = (hd['admission_ward'] == 'ICU').astype(int)
hd['discharged_against_advice'] = (hd['outcome_during_hospitalization'] == 'DischargeAgainstOrder').astype(int)
hd['discharged_to_facility'] = (hd['destinationdischarge'] == 'HealthcareFacility').astype(int)
hd['oxygen_therapy'] = (hd['oxygen_inhalation'] == 'OxygenTherapy').astype(int)
print('ventilated:', hd['ventilated'].sum(), '| died in hospital:', hd['died_in_hospital'].sum(), '| against advice:', hd['discharged_against_advice'].sum())

labs = t['labs'].copy()
REDUNDANT = ['globulin','white_globulin_ratio','total_bilirubin','neutrophil_count','creatine_kinase_isoenzyme_to_creatine_kinase',
             'hydroxybutyrate_dehydrogenase_to_lactate_dehydrogenase','map_value','potassium_ion','sodium_ion','chloride_ion','hematocrit',
             'total_hemoglobin','hematocrit_blood_gas','total_carbon_dioxide','measured_bicarbonate','measured_residual_base','oxyhemoglobin']
BLOOD_GAS = ['ph','standard_residual_base','standard_bicarbonate','partial_pressure_of_carbon_dioxide','methemoglobin','reduced_hemoglobin','glucose_blood_gas',
             'lactate','carboxyhemoglobin','body_temperature_blood_gas','oxygen_saturation','partial_oxygen_pressure','anion_gap','free_calcium']
SKEWED = ['brain_natriuretic_peptide','high_sensitivity_troponin','creatinine_enzymatic_method','urea','d_dimer','high_sensitivity_protein',
          'glutamic_pyruvic_transaminase','glutamic_oxaloacetic_transaminase','total_bilirubin','creatine_kinase','lactate_dehydrogenase','cystatin']
num_cols = [c for c in labs.columns if c != 'inpatient_number']
labs[num_cols] = labs[num_cols].apply(pd.to_numeric, errors='coerce').astype(float)
missing_pct = labs[num_cols].isna().mean()
model_cols = [c for c in num_cols if c not in REDUNDANT + BLOOD_GAS and missing_pct[c] < 0.25]
for c in SKEWED: labs['log_' + c] = np.log1p(labs[c])
labs['bnp_at_ceiling'] = (labs['brain_natriuretic_peptide'] >= 5000).astype(int)
labs['blood_gas_done'] = labs['ph'].notna().astype(int)
labs['ckd_stage'] = pd.cut(labs['glomerular_filtration_rate'], [0,15,30,45,60,90,1000], right=False, labels=['G5','G4','G3b','G3a','G2','G1'])
flag = lambda s, cond: np.where(s.isna(), np.nan, cond.astype(float))
labs['hyponatremia'] = flag(labs['sodium'], labs['sodium'] < 135)
labs['hyperkalemia'] = flag(labs['potassium'], labs['potassium'] > 5.5)
labs['hypokalemia'] = flag(labs['potassium'], labs['potassium'] < 3.5)
labs['hypoalbuminemia'] = flag(labs['albumin'], labs['albumin'] < 35)
labs['troponin_elevated'] = flag(labs['high_sensitivity_troponin'], labs['high_sensitivity_troponin'] > 14)   # pg/mL
labs['tachycardia'] = flag(labs['pulse'], labs['pulse'] > 100)
labs['hypotension'] = flag(labs['systolic_blood_pressure'], labs['systolic_blood_pressure'] < 90)
labs['bnp_quartile'] = pd.qcut(labs['brain_natriuretic_peptide'], 4, labels=['Q1','Q2','Q3','Q4'])
lm = {'model_columns': model_cols, 'redundant': REDUNDANT, 'skewed_logged': SKEWED}
bnp = labs['brain_natriuretic_peptide']
print(f'model columns: {len(model_cols)} | BNP measured: {bnp.notna().sum()} (median {bnp.median():.0f}, max {bnp.max():.0f} pg/mL) | at 5000 ceiling: {labs.bnp_at_ceiling.sum()}')

ph = t['hist'].copy()
ph['n_major_comorb'] = ph[['diabetes','moderate_to_severe_chronic_kidney_disease','chronic_obstructive_pulmonary_disease','cerebrovascular_disease','dementia']].sum(axis=1)
ph['cci_group'] = pd.cut(ph['cci_score'], [-1, 1, 2, 3, 20], labels=['0-1','2','3','4+'])
rs = t['resp'].copy()
rs['consciousness_conflict'] = ((rs['consciousness'] == 'Clear') & (rs['gcs'] < 15)).astype(int)   # label says Clear but GCS < 15
rs['impaired_consciousness'] = (rs['gcs'] < 15).astype(int)                                        # GCS is the source of truth
rs['gcs_severity'] = pd.cut(rs['gcs'], [0, 8, 12, 14, 15], labels=['Severe','Moderate','Mild','Normal'])
print('cci_score missing (kept as NaN by Team05):', ph['cci_score'].isna().sum(), '| impaired by GCS:', rs['impaired_consciousness'].sum(), '| label conflicts:', rs['consciousness_conflict'].sum())

rx = t['rx'].copy()
rx['drug'] = rx['drug_name'].str.strip().str.lower()        # drug names rebuilt from the drug_ column names
CLASS = {
 'furosemide injection':('loop_diuretic','IV'), 'furosemide tablet':('loop_diuretic','oral'), 'torasemide tablet':('loop_diuretic','oral'),
 'hydrochlorothiazide tablet':('thiazide','oral'), 'spironolactone tablet':('mra','oral'),
 'digoxin tablet':('digitalis','oral'), 'deslanoside injection':('digitalis','IV'),
 'milrinone injection':('iv_inotrope','IV'), 'dobutamine hydrochloride injection':('iv_inotrope','IV'), 'isoprenaline hydrochloride injection':('iv_inotrope','IV'),
 'metoprolol succinate sustained-release tablet':('beta_blocker','oral'), 'metoprolol tartrate injection':('beta_blocker','IV'),
 'benazepril hydrochloride tablet':('acei_arb','oral'), 'valsartan dispersible tablet':('acei_arb','oral'),
 'aspirin enteric-coated tablet':('antiplatelet','oral'), 'clopidogrel hydrogen sulphate tablet':('antiplatelet','oral'),
 'atorvastatin calcium tablet':('statin','oral'),
 'warfarin sodium tablet':('anticoagulant','oral'), 'heparin sodium injection':('anticoagulant','IV'), 'enoxaparin sodium injection':('anticoagulant','IV'),
 'isosorbide mononitrate sustained release tablet':('nitrate','oral'), 'nitroglycerin injection':('nitrate','IV'),
 'meglumine adenosine cyclophosphate for injection':('metabolic_support','IV'),
 'shenfu injection':('tcm','IV'), 'sulfotanshinone sodium injection':('tcm','IV')}
assert set(rx['drug']) <= set(CLASS), set(rx['drug']) - set(CLASS)
rx['drug_class'] = rx['drug'].map(lambda d: CLASS[d][0]); rx['route'] = rx['drug'].map(lambda d: CLASS[d][1])
slug = lambda s: 'rx_' + s.replace(' ', '_').replace('-', '_')
drug_wide = pd.crosstab(rx['inpatient_number'], rx['drug']).clip(upper=1); drug_wide.columns = [slug(c) for c in drug_wide.columns]
class_wide = pd.crosstab(rx['inpatient_number'], rx['drug_class']).clip(upper=1); class_wide.columns = ['cls_' + c for c in class_wide.columns]
wide = drug_wide.join(class_wide)
wide['n_drugs'] = rx.groupby('inpatient_number').size()
wide['n_iv_drugs'] = rx[rx['route'] == 'IV'].groupby('inpatient_number').size().reindex(wide.index).fillna(0)
wide['iv_loop_diuretic'] = drug_wide['rx_furosemide_injection']
all_ids = t['hosp']['inpatient_number']
print('Patients without any prescription (kept, all flags 0):', sorted(set(all_ids) - set(wide.index)))
wide = wide.reindex(all_ids.values).fillna(0).astype(int)
wide['has_rx'] = (wide['n_drugs'] > 0).astype(int)
wide['gdmt_count'] = wide['rx_metoprolol_succinate_sustained_release_tablet'] + wide['cls_acei_arb'] + wide['cls_mra']
wide['gdmt_all3'] = (wide['gdmt_count'] == 3).astype(int)
wide['rescue_therapy'] = wide['cls_iv_inotrope']
wide = wide.rename_axis('inpatient_number').reset_index()
print('all 3 GDMT pillars: {:.1f}%'.format(wide['gdmt_all3'].mean() * 100))

# ---- 3. Merge into one analysis table, add cross-file fields ----
T = {'demo': demo, 'cardiac': cc, 'hosp': hd, 'labs': labs, 'hist': ph, 'resp': rs, 'rx': wide}

id_sets = {k: set(d['inpatient_number']) for k, d in T.items()}
base = id_sets['hosp']
print({k: (len(v), len(v ^ base)) for k, v in id_sets.items()}, '← (n IDs, n differing from hosp)')
df = T['hosp']
for k in ['demo','cardiac','labs','hist','resp','rx']:
    overlap = (set(df.columns) & set(T[k].columns)) - {'inpatient_number'}
    assert not overlap, overlap
    df = df.merge(T[k], on='inpatient_number', how='inner', validate='one_to_one')
print('Combined shape:', df.shape)

thr = np.where(df['gender'] == 'Male', 130, 120)
df['anemia'] = np.where(df['hemoglobin'].isna(), np.nan, (df['hemoglobin'] < thr).astype(float))
df['anemia_severity'] = pd.cut(df['hemoglobin'], [0, 80, 110, 1e9], labels=['Severe (<80)','Moderate (80-109)','Mild/None (≥110)'])
print(df.groupby('gender')['anemia'].mean().mul(100).round(1).rename('anaemia %'))

print(df.groupby('gender')['height'].median().rename('median height (m)'))
g = df.dropna(subset=['glucose_blood_gas'])
print('\nBlood-gas glucose ≥ 11.1 mmol/L by diabetes flag (%):')
print(g.groupby('diabetes')['glucose_blood_gas'].apply(lambda s: round((s >= 11.1).mean()*100, 1)))
print('\nDied in hospital AND readmitted within 6 m:', ((df['died_in_hospital']==1) & (df['re_admission_within_6_months']==1)).sum())
print('Ventilated % by impaired consciousness:', df.groupby('impaired_consciousness')['ventilated'].mean().mul(100).round(1).to_dict())

df['hyperglycemia_nondiabetic'] = ((df['diabetes'] == 0) & (df['glucose_blood_gas'] >= 11.1)).astype(int)
print('Flagged:', df['hyperglycemia_nondiabetic'].sum())

OUTCOMES = ['outcome_during_hospitalization','destinationdischarge','death_within_28_days','re_admission_within_28_days','death_within_3_months',
            're_admission_within_3_months','death_within_6_months','re_admission_within_6_months','time_of_death__days_from_admission',
            'readmission_time_days_from_admission','return_to_emergency_department_within_6_months','died_in_hospital','outcome_conflict',
            'death_or_readm_6m','discharged_against_advice','discharged_to_facility']
labs_meta = lm
FEATURES = {
 'demography': ['age_mid','gender','bmi','occupation'],
 'cardiac': ['nyha_cardiac_function_classification','killip_grade','myocardial_infarction','peripheral_vascular_disease','type_of_heart_failure','lvef','left_ventricular_end_diastolic_diameter_lv'],
 'labs': labs_meta['model_columns'],
 'history': ['cerebrovascular_disease','dementia','chronic_obstructive_pulmonary_disease','peptic_ulcer_disease','diabetes','moderate_to_severe_chronic_kidney_disease','solid_tumor','liver_disease','cci_score','type_ii_respiratory_failure'],
 'responsiveness': ['gcs'],
 'admission': ['admission_way','admission_ward','oxygen_inhalation'],
 'during_stay': ['dischargeday','ventilated','discharge_department','visit_times'],
 'prescriptions': [c for c in df.columns if c.startswith('cls_')] + ['n_drugs','n_iv_drugs','iv_loop_diuretic','gdmt_count'],
}
FEATURES['labs'] = [c for c in FEATURES['labs'] if c != 'glomerular_filtration_rate'] + ['glomerular_filtration_rate']
fs = {'outcomes': OUTCOMES, 'features': FEATURES}
print({k: len(v) for k, v in FEATURES.items()})

# ---- 4. Final analysis table (same types as the saved file used by notebooks 09-11) ----
df = pd.read_csv(io.StringIO(df.to_csv(index=False)), parse_dates=['admission_date'])
for c, o in ORDERS.items():
    if c in df: df[c] = pd.Categorical(df[c], categories=o, ordered=True)
plt.rcParams.update({'figure.dpi': 90, 'axes.spines.top': False, 'axes.spines.right': False})
rt = rate_table

# ---- 5. Definitions used by the predictive questions (H1-H50) ----
logmap = lambda c: 'log_' + c if c in lm['skewed_logged'] else c
DEMO_N, DEMO_C = ['age_mid','bmi'], ['gender','occupation']
CARD_N, CARD_C = ['nyha_cardiac_function_classification','killip_grade','myocardial_infarction','peripheral_vascular_disease'], ['type_of_heart_failure']
ECHO = ['lvef','left_ventricular_end_diastolic_diameter_lv','ea','tricuspid_valve_return_pressure']
LAB_ALL = [logmap(c) for c in lm['model_columns']] + ['blood_gas_done','bnp_at_ceiling']
LAB_CORE = ['log_brain_natriuretic_peptide','log_high_sensitivity_troponin','glomerular_filtration_rate','log_urea','sodium','potassium','hemoglobin','albumin',
            'lymphocyte_count','neutrophil_ratio','coefficient_of_variation_of_red_blood_cell_distribution_width','log_d_dimer','uric_acid','pulse','systolic_blood_pressure',
            'respiration','carbon_dioxide_binding_capacity','blood_gas_done']
HIST = ['cerebrovascular_disease','dementia','chronic_obstructive_pulmonary_disease','peptic_ulcer_disease','diabetes','moderate_to_severe_chronic_kidney_disease',
        'solid_tumor','liver_disease','cci_score','type_ii_respiratory_failure']
RESP = ['gcs','eye_opening','verbal_response','movement']
ADM_C = ['admission_way','admission_ward','oxygen_inhalation']; ADM_N = ['fio2']
STAY_N, STAY_C = ['log_los','ventilated'], ['discharge_department']
RX = [c for c in df.columns if c.startswith('cls_')] + ['n_drugs','n_iv_drugs','iv_loop_diuretic','gdmt_count']
ADMIT_N = DEMO_N + CARD_N + ECHO + LAB_CORE + HIST + ['gcs'] + ADM_N
ADMIT_C = DEMO_C + CARD_C + ADM_C
df['hfref'] = np.where(df['lvef'].isna(), np.nan, (df['lvef'] < 40).astype(float))
df['survived_hosp'] = ((df['died_in_hospital']==0) & (df['death_within_6_months']==0)).astype(int)
RESULTS = []
LABEL = {'death_within_28_days':'28-day death','death_within_3_months':'3-month death','death_within_6_months':'6-month death','re_admission_within_28_days':'28-day readmission',
         're_admission_within_3_months':'3-month readmission','re_admission_within_6_months':'6-month readmission','prolonged_los':'stay ≥ 14 days','death_or_readm_6m':'death or readmission (6 m)',
         'discharged_to_facility':'discharge to a care facility','discharged_against_advice':'discharge against advice','ventilated':'need for ventilation','cls_iv_inotrope':'IV inotrope use',
         'killip_high':'Killip 3–4','nyha_iv':'NYHA IV','hfref':'HFrEF (LVEF < 40%)','re_admission_within_6_months ':'6-month readmission'}

def direction(top):
    out = []
    for k, v in top.head(5).items():
        out.append(f"`{k}` ({'↑ risk' if v > 0 else '↓ risk'})")
    return ', '.join(out)

def run_h(hid, outcome, num, cat=(), kind='logreg', data=None, compare_hgb=False, fig=False):
    d = df if data is None else data
    num = [c for c in num if c != outcome]; cat = [c for c in cat if c != outcome]
    scr = screen_markers(d, [c for c in num if d[c].nunique() > 2][:40], outcome).head(6) if any(d[c].nunique() > 2 for c in num) else None
    res = evaluate_classifier(d, outcome, num, cat, kind=kind)
    if compare_hgb:
        alt = evaluate_classifier(d, outcome, num, cat, kind='hgb')
        res['alt_model'] = 'HGB' ; res['alt_AUC'] = alt['AUC']; res['alt_CI'] = alt['AUC_95CI']
        if alt['AUC'] > res['AUC'] + 0.01: alt['alt_model'] = 'logistic'; alt['alt_AUC'] = res['AUC']; alt['alt_CI'] = res['AUC_95CI']; res = alt
    top = top_features(d, outcome, num, cat, k=8)
    q = risk_deciles(res['_prob'], res['_y'])
    res.update({'id': hid, 'n_features': len(num) + len(cat), 'top': top, 'quint': q, 'lift_top': round(q['observed_rate'].iloc[-1] / max(q['observed_rate'].iloc[0], 0.1), 1),
                'top_q_share': round(res['_y'][res['_prob'] >= np.quantile(res['_prob'], .8)].sum() / res['_y'].sum() * 100, 0)})
    res['verdict'] = hypothesis_verdict(res)
    RESULTS.append({k: v for k, v in res.items() if not k.startswith('_') and k not in ('top','quint')} | {'top_markers': ', '.join(top.index[:5])})
    if fig:
        from sklearn.metrics import roc_curve
        f, ax = plt.subplots(1, 2, figsize=(10, 3.4)); fpr, tpr, _ = roc_curve(res['_y'], res['_prob'])
        ax[0].plot(fpr, tpr, color=PALETTE['main']); ax[0].plot([0,1],[0,1],'k--', lw=1); ax[0].set_title(f"{hid} ROC — AUC {res['AUC']}"); ax[0].set_xlabel('1 − specificity'); ax[0].set_ylabel('sensitivity')
        ax[1].bar(q.index.astype(str), q['observed_rate'], color=PALETTE['blue']); ax[1].set_title('Observed rate by predicted-risk quintile'); ax[1].set_ylabel('%')
        plt.tight_layout(); savefig(f, hid + '_model')
    print(res['verdict']); print({k: res[k] for k in ['n','events','prevalence_%','AUC','AUC_95CI','PR_AUC','baseline_PR_AUC','balanced_acc','sensitivity','specificity']})
    if compare_hgb: print('Model used:', res['model'], '| other model AUC:', res['alt_AUC'])
    if scr is not None: print('\nMarker screen (Mann-Whitney):'); print(scr.to_string(index=False))
    print('\nStandardised logistic coefficients:'); print(top.to_string()); print('\nRisk quintiles:'); print(q.T.to_string())
    return res
print('Setup complete: analysis table', df.shape)
```

### H38 · Can admission data predict the actual number of days in hospital?

```python
from scipy.stats import pearsonr
res_H38 = evaluate_regressor(df, 'log_los', ADMIT_N, ADMIT_C)
pred_days = np.expm1(res_H38['_pred']); true_days = np.expm1(res_H38['_y'])
mae_days = np.mean(np.abs(pred_days - true_days)); mae_base = np.mean(np.abs(np.median(true_days) - true_days))
r38, p38 = pearsonr(res_H38['_pred'], res_H38['_y'])
print({k: v for k, v in res_H38.items() if not k.startswith('_')}, '| MAE days', round(mae_days,2), 'vs median-guess', round(mae_base,2), '| r', round(r38,3), p38)
RESULTS.append({'id':'H38','outcome':'log LOS (regression)','model':'RF regressor','n':res_H38['n'],'AUC':None,'R2':res_H38['R2'],'top_markers':'kidney, sodium, comorbidity'})
```

### H50 · Does the readmission model generalise to a future year (train 2016–2018, test 2019)?

```python
from sklearn.metrics import roc_auc_score
from scipy.stats import mannwhitneyu
out50 = {}
for outc in ['re_admission_within_6_months', 'death_within_6_months', 'death_within_28_days']:
    tr, te = df[df['adm_year'] <= 2018], df[df['adm_year'] == 2019]
    m = make_model('logreg', ADMIT_N, ADMIT_C).fit(tr[ADMIT_N + ADMIT_C], tr[outc].astype(int))
    p = m.predict_proba(te[ADMIT_N + ADMIT_C])[:, 1]; y = te[outc].astype(int).values
    lo, hi = auc_ci(p, y); out50[outc] = {'train_n': len(tr), 'test_n': len(te), 'test_events': int(y.sum()), 'AUC_2019': round(roc_auc_score(y, p), 3), 'CI': f'{lo:.3f}-{hi:.3f}',
                                         'p': mannwhitneyu(p[y==1], p[y==0], alternative='greater').pvalue}
t50 = pd.DataFrame(out50).T; RESULTS.append({'id':'H50','outcome':'6m readmission (temporal)','model':'logreg','n':int(t50.iloc[0]['test_n']),'AUC':t50.loc['re_admission_within_6_months','AUC_2019'],'top_markers':'temporal validation'})
t50
```
