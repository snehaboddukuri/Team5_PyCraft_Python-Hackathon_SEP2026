Step 1 - Load the Data

```python
import pandas as pd
import numpy as np

cardiac = pd.read_csv("cardiac_complications.csv")

print(cardiac.shape)
print(cardiac.info())
```

Step 2 - Check Missing Values

```python
missing = cardiac.isnull().sum()

missing_percent = (
    cardiac.isnull().sum()
    / len(cardiac)
    * 100
).round(2)

missing_df = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": missing_percent
})

missing_df.sort_values(
    by="Missing %",
    ascending=False
)
```

Save Missing Value Result

```python
missing_df.to_csv(
    "cardiac_missing_report.csv"
)
```

Step 3 - Check For Duplicate Record

```python
cardiac.duplicated().sum()
```

Step 4 - Convert Categorical variable

```python
categorical_cols = [
    'type_of_heart_failure'
]

for col in categorical_cols:
    cardiac[col] = cardiac[col].astype('category')
```
