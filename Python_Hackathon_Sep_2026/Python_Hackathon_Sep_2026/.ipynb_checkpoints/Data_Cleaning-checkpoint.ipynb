## 1. Cleaning patient_precriptions table

## Reading csv file
We start by loading the raw patient_precriptions data into a DataFrame so we can inspect and clean it.

```python
#Importing all the Necessary Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

import warnings
warnings.simplefilter("ignore", UserWarning)

#import os
#os.chdir("Python_Hackathon_SEP_2026")

#Reading the file 
df_precriptions=pd.read_csv("cardiac_failure/patient_precriptions.csv")
```

## Renaming columns
Raw column names like inpatient_number . We rename them to patient_id so the rest of the notebook (and anyone reading it later) can understand what each column actually means, and so column names match across all our tables.

```python
# Mapping old column names to new column names
df_precriptions = df_precriptions.rename(columns={"inpatient_number": "patient_id"})

```

```python
df_precriptions
```

## Checking and Standardizing column values

Text columns often contain inconsistent formatting — extra spaces, mismatched capitalization, or missing entries . We strip whitespace and standardize capitalization for `drug_name` so identical values are always treated as the same category later on.

```python
# 2. Clean text column formatting
# Strip accidental blank spaces from sides and capitalize every word perfectly
df_precriptions["drug_name"] = df_precriptions["drug_name"].str.strip().str.title()
```

## Check for data types


```python
# Get a quick summary of columns, data types, and missing values
print(df_precriptions.info())
```

## Check for duplicate records

Duplicate rows can silently inflate patient counts and bias any analysis or model built on this data. We count exact duplicate rows and inspect them, so we can decide whether they're true duplicates (safe to drop) or coincidentally identical valid records.

```python
# Count how many total exact duplicate rows exist
num_duplicates = df_precriptions.duplicated().sum()
print("Total exact duplicate rows:", num_duplicates)

# Display the duplicate rows if any exist
if num_duplicates > 0:
    print("\nDuplicate Rows:")
    print(df_precriptions[df_precriptions.duplicated(keep=False)])
else:
    print("No exact duplicate rows found.")
```

```python
# 3. Deduplicate exact duplicate transaction rows
df_precriptions = df_precriptions.drop_duplicates()
```

```python
# 5. Export to a clean CSV file
df_precriptions.to_csv("cleaned_inpatient_drugs.csv", index=False)

# Preview the formatted result
print(df_precriptions.head(10))
```

## Checking for Missing values

For cholinesterase, its 100% empty , so we will remove this column in our cleaning method.                                               

```python

with pd.option_context("display.max_rows", None):
    print(df_labs.isna().sum())

    #print(df_labs["CHE"].isna().sum())
```

 ## clean_labs_pipeline method
 This method Standardizes, cleans, and normalizes clinical lab results and vital signs. Resolves inconsistent naming conventions, eliminates non-numeric entries, drops unphysiological outliers (like zero-vitals and extreme system overflows), casts whole numbers to nullable integer formats, and dynamically normalizes variant assay units like Troponin.

```python
def clean_labs_pipeline(dataframe, mapping_dict, ranges_dict):
    cleaned_df = dataframe.copy()

    # Step A: Standardize column headers
    cleaned_df = cleaned_df.rename(columns=mapping_dict)

    # Step B: Coerce to Numeric Data Types
    for col in mapping_dict.values():
        if col in cleaned_df.columns:
            cleaned_df[col] = pd.to_numeric(cleaned_df[col], errors="coerce")

    # Step C: Handle Impossible Vitals (Zeros -> NaN)
    vitals = ["pulse", "resp", "sbp", "dbp"]
    for v in vitals:
        if v in cleaned_df.columns:
            cleaned_df.loc[cleaned_df[v] == 0, v] = np.nan

      # Step D: Force Vitals, Hemoglobin, and Platelets to be Whole Numbers
    # This strips the `.0` decimal artifact while safely preserving empty cells"
    whole_number_cols = ["pulse", "resp", "sbp", "dbp", "uric_acid","MCHC","hemoglobin", "platelet","glutamic_oxaloacetic_transaminase","creatine_kinase","lactate_dehydrogenase","glutamyltranspeptidase","glutamic_pyruvic_transaminase","hydroxybutyrate_dehydrogenase","ALP","pCO2","HCT_BG","Cl_ion","temp_BG","sO2","pO2","tHb"]
    for col in whole_number_cols:
        if col in cleaned_df.columns:
            # Round first to handle float representation discrepancies, then cast to Int64
            cleaned_df[col] = cleaned_df[col].round(0).astype("Int64")

    # Step E: Fix Units to Match Dictionary (Fractions -> Percentages)
    percentage_cols = ["monocyte_ratio", "eosinophil_ratio", "basophil_ratio", "HCT"]
    for pc in percentage_cols:
        if pc in cleaned_df.columns:
            cleaned_df[pc] = (cleaned_df[pc] * 100).round(4)

        # Step F: Dynamic Troponin Unit Normalization Logic
    if "high_sensitivity_troponin" in cleaned_df.columns:
        if cleaned_df["high_sensitivity_troponin"].median() < 1.0:
            print("Converting troponin from ng/mL to pg/mL (ng/L)...")
            cleaned_df["high_sensitivity_troponin"] = (cleaned_df["high_sensitivity_troponin"] * 1000).round(1)
        else:
            print("Troponin is already in pg/mL. Keeping raw values.")
            cleaned_df["high_sensitivity_troponin"] = cleaned_df["high_sensitivity_troponin"].round(1)

    # Step G: Value Cleanup Loop (Removes extreme system overflows like 5000 overrides)
    for col, (min_val, max_val) in ranges_dict.items():
        if col in cleaned_df.columns:
            # Drop entries that are over 10x the reference max threshold
            cleaned_df.loc[cleaned_df[col] > (max_val * 10), col] = np.nan

   # Step H: Rounding down
            cleaned_df["MAP"] = cleaned_df["MAP"].round(4)

            #df_labs = df_labs.drop(columns=["cholinesterase"])
           # cleaned_df["cholinesterase"]=cleaned_df.drop(columns=["cholinesterase"])

         # Step I: Drop unnecessary or corrupted columns
    if "cholinesterase" in cleaned_df.columns:
        cleaned_df = cleaned_df.drop(columns=["cholinesterase"])
        
    return cleaned_df
   
```

```python
# 5. Run the pipeline execution turn
df_labs = clean_labs_pipeline(df_labs, COLUMN_MAPPING, REFERENCE_BOUNDS)
```
