```python
# ============================================================
# HEART FAILURE COHORT ANALYSIS
# Questions 1–7
# ============================================================

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from scipy import stats

import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.proportion import proportions_ztest

from sklearn.preprocessing import SplineTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# ------------------------------------------------------------
# 0. LOAD DATA
# ------------------------------------------------------------


df = pd.read_csv("Team05_PyCraft_Cleaned_data/wide_merged_table.csv")

print(df.shape)
print(df.head())
print(df.info())


# ============================================================
# COLUMN MAPPING
# Change these names to match your dataset
# ============================================================

COLS = {
    "age": "age",
    "sex": "sex",
    "diabetes": "diabetes",
    "eGFR": "eGFR",
    "potassium": "potassium",

    "spironolactone": "spironolactone",
    "sglt2i": "sglt2i",
    "ace_arb_arni": "ace_arb_arni",
    "beta_blocker": "beta_blocker",

    "department": "department",
    "occupation": "occupation",
    "echo": "echo",

    "readmission_6m": "readmission_6m",
    "death_6m": "death_6m",

    # Optional baseline covariates
    "creatinine": "creatinine",
    "heart_rate": "heart_rate",
    "sbp": "sbp",
    "hemoglobin": "hemoglobin",
}


# ============================================================
# 1. BASIC DATA CLEANING
# ============================================================

df = df.copy()

# Convert numeric variables
numeric_cols = [
    COLS["age"],
    COLS["eGFR"],
    COLS["potassium"],
    COLS["creatinine"],
    COLS["heart_rate"],
    COLS["sbp"],
    COLS["hemoglobin"],
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")


# Convert binary variables to 0/1 where appropriate
binary_cols = [
    COLS["diabetes"],
    COLS["spironolactone"],
    COLS["sglt2i"],
    COLS["ace_arb_arni"],
    COLS["beta_blocker"],
    COLS["echo"],
    COLS["readmission_6m"],
    COLS["death_6m"],
]

for col in binary_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")


# ============================================================
# HELPER FUNCTION
# ============================================================

def logistic_model(formula, data=df):
    """
    Logistic regression with odds ratios and 95% CI.
    """
    model = smf.logit(formula, data=data).fit(disp=False)

    results = pd.DataFrame({
        "OR": np.exp(model.params),
        "CI_lower": np.exp(model.conf_int()[0]),
        "CI_upper": np.exp(model.conf_int()[1]),
        "p_value": model.pvalues
    })

    return model, results


# ============================================================
# 1. DIABETES + HF MEDICATIONS AND AKI
# ============================================================

# Requires an AKI variable:
# AKI = 1 if AKI occurred during follow-up

# Example:
# df["AKI"] = ...

meds = [
    COLS["sglt2i"],
    COLS["ace_arb_arni"],
    COLS["beta_blocker"],
    COLS["spironolactone"]
]

print("\nMedication use:")
print(df[meds].mean())


# Number of HF medication classes
df["n_gdmt"] = df[meds].sum(axis=1)

print("\nNumber of medication classes:")
print(df["n_gdmt"].value_counts().sort_index())


# AKI among diabetic patients
diabetic = df[df[COLS["diabetes"]] == 1].copy()

if "AKI" in diabetic.columns:

    formula = (
        "AKI ~ "
        f"{COLS['sglt2i']} + "
        f"{COLS['ace_arb_arni']} + "
        f"{COLS['beta_blocker']} + "
        f"{COLS['spironolactone']} + "
        f"{COLS['eGFR']} + "
        f"{COLS['age']} + "
        f"{COLS['potassium']}"
    )

    model_1, results_1 = logistic_model(formula, diabetic)

    print("\n=== AKI model among diabetic patients ===")
    print(results_1)


# Medication combinations
df["GDMT_combo"] = (
    df[meds]
    .astype(str)
    .agg("-".join, axis=1)
)

if "AKI" in diabetic.columns:

    combo_table = (
        diabetic
        .groupby("GDMT_combo")["AKI"]
        .agg(["count", "mean", "sum"])
        .sort_values("mean")
    )

    print("\n=== AKI by medication combination ===")
    print(combo_table)


# ============================================================
# 2. CARDIOLOGY VS GENERAL WARD
# ============================================================

print("\n=== Readmission by discharge department ===")

department_table = (
    df.groupby(COLS["department"])[COLS["readmission_6m"]]
      .agg(["count", "mean", "sum"])
)

print(department_table)


# Crude logistic regression
formula = (
    f"{COLS['readmission_6m']} ~ C({COLS['department']})"
)

model_2a, results_2a = logistic_model(formula)

print("\n=== Crude department model ===")
print(results_2a)


# Adjusted model
formula = (
    f"{COLS['readmission_6m']} ~ "
    f"C({COLS['department']}) + "
    f"{COLS['age']} + "
    f"{COLS['eGFR']} + "
    f"{COLS['potassium']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_2b, results_2b = logistic_model(formula)

print("\n=== Adjusted department model ===")
print(results_2b)


# ============================================================
# 3. AGE AND OUTCOMES
# ============================================================

# Create clinically interpretable age groups

df["age_group"] = pd.cut(
    df[COLS["age"]],
    bins=[0, 64, 79, np.inf],
    labels=["<65", "65-79", "80+"]
)

print("\n=== Readmission by age group ===")

age_table = (
    df.groupby("age_group", observed=False)
      .agg(
          n=(COLS["readmission_6m"], "size"),
          readmission=(COLS["readmission_6m"], "mean"),
          mortality=(COLS["death_6m"], "mean")
      )
)

print(age_table)


# Readmission model
formula = (
    f"{COLS['readmission_6m']} ~ "
    f"C(age_group) + "
    f"{COLS['eGFR']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']} + "
    f"{COLS['department'] if False else COLS['department']}"
)

model_3a, results_3a = logistic_model(formula)

print("\n=== Age group and readmission ===")
print(results_3a)


# Mortality model
formula = (
    f"{COLS['death_6m']} ~ "
    f"C(age_group) + "
    f"{COLS['eGFR']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_3b, results_3b = logistic_model(formula)

print("\n=== Age group and mortality ===")
print(results_3b)


# ============================================================
# 4. OCCUPATION / POSSIBLE RURAL-URBAN PROXY
# ============================================================

print("\n=== Occupation distribution ===")
print(df[COLS["occupation"]].value_counts(dropna=False))


# Crude association
formula = (
    f"{COLS['readmission_6m']} ~ "
    f"C({COLS['occupation']})"
)

model_4a, results_4a = logistic_model(formula)

print("\n=== Occupation and readmission ===")
print(results_4a)


# Adjusted
formula = (
    f"{COLS['readmission_6m']} ~ "
    f"C({COLS['occupation']}) + "
    f"{COLS['age']} + "
    f"{COLS['eGFR']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']} + "
    f"C({COLS['department']})"
)

model_4b, results_4b = logistic_model(formula)

print("\n=== Adjusted occupation model ===")
print(results_4b)


# ============================================================
# 5. NO ECHO RECORDED
# ============================================================

print("\n=== Echo documentation ===")

echo_table = (
    df.groupby(COLS["echo"])
      .agg(
          n=(COLS["readmission_6m"], "size"),
          readmission=(COLS["readmission_6m"], "mean"),
          mortality=(COLS["death_6m"], "mean")
      )
)

print(echo_table)


# Readmission
formula = (
    f"{COLS['readmission_6m']} ~ "
    f"{COLS['echo']} + "
    f"{COLS['age']} + "
    f"{COLS['eGFR']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']} + "
    f"C({COLS['department']})"
)

model_5a, results_5a = logistic_model(formula)

print("\n=== Echo documentation and readmission ===")
print(results_5a)


# Mortality
formula = (
    f"{COLS['death_6m']} ~ "
    f"{COLS['echo']} + "
    f"{COLS['age']} + "
    f"{COLS['eGFR']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_5b, results_5b = logistic_model(formula)

print("\n=== Echo documentation and mortality ===")
print(results_5b)


# ============================================================
# 6. eGFR STAGE AND OUTCOMES
# ============================================================

# KDIGO-style categories
# G1 >=90
# G2 60-89
# G3a 45-59
# G3b 30-44
# G4 15-29
# G5 <15

df["eGFR_stage"] = pd.cut(
    df[COLS["eGFR"]],
    bins=[-np.inf, 15, 30, 45, 60, 90, np.inf],
    labels=["G5", "G4", "G3b", "G3a", "G2", "G1"],
    right=False
)

print("\n=== eGFR stage ===")
print(df["eGFR_stage"].value_counts().sort_index())


# Readmission by stage
egfr_table = (
    df.groupby("eGFR_stage", observed=False)
      .agg(
          n=(COLS["readmission_6m"], "size"),
          readmission=(COLS["readmission_6m"], "mean"),
          mortality=(COLS["death_6m"], "mean")
      )
)

print("\n=== Outcomes by eGFR stage ===")
print(egfr_table)


# Categorical model
formula = (
    f"{COLS['readmission_6m']} ~ "
    f"C(eGFR_stage) + "
    f"{COLS['age']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_6a, results_6a = logistic_model(formula)

print("\n=== eGFR stage and readmission ===")
print(results_6a)


# Mortality
formula = (
    f"{COLS['death_6m']} ~ "
    f"C(eGFR_stage) + "
    f"{COLS['age']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_6b, results_6b = logistic_model(formula)

print("\n=== eGFR stage and mortality ===")
print(results_6b)


# ------------------------------------------------------------
# Test for NON-LINEAR relationship using restricted cubic
# splines
# ------------------------------------------------------------

# Simple spline using patsy
from patsy import bs

formula = (
    f"{COLS['readmission_6m']} ~ "
    f"bs({COLS['eGFR']}, df=4) + "
    f"{COLS['age']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_6c = smf.logit(formula, data=df).fit(disp=False)

print("\n=== Nonlinear eGFR model: readmission ===")
print(model_6c.summary())


formula = (
    f"{COLS['death_6m']} ~ "
    f"bs({COLS['eGFR']}, df=4) + "
    f"{COLS['age']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_6d = smf.logit(formula, data=df).fit(disp=False)

print("\n=== Nonlinear eGFR model: mortality ===")
print(model_6d.summary())


# ============================================================
# 7. POTASSIUM + SPIRONOLACTONE
# ============================================================

# Categorize potassium
# Modify thresholds according to your clinical protocol.

df["K_group"] = pd.cut(
    df[COLS["potassium"]],
    bins=[-np.inf, 3.5, 4.0, 4.5, 5.0, 5.5, np.inf],
    labels=[
        "<3.5",
        "3.5-4.0",
        "4.0-4.5",
        "4.5-5.0",
        "5.0-5.5",
        ">=5.5"
    ]
)

print("\n=== Potassium distribution ===")
print(df["K_group"].value_counts().sort_index())


# Mortality by potassium
K_table = (
    df.groupby("K_group", observed=False)
      .agg(
          n=(COLS["death_6m"], "size"),
          mortality=(COLS["death_6m"], "mean"),
          readmission=(COLS["readmission_6m"], "mean")
      )
)

print("\n=== Outcomes by potassium ===")
print(K_table)


# Potassium + MRA interaction
formula = (
    f"{COLS['death_6m']} ~ "
    f"C(K_group) * {COLS['spironolactone']} + "
    f"{COLS['age']} + "
    f"{COLS['eGFR']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_7a, results_7a = logistic_model(formula)

print("\n=== Potassium × spironolactone: mortality ===")
print(results_7a)


# Readmission interaction
formula = (
    f"{COLS['readmission_6m']} ~ "
    f"C(K_group) * {COLS['spironolactone']} + "
    f"{COLS['age']} + "
    f"{COLS['eGFR']} + "
    f"{COLS['diabetes']} + "
    f"{COLS['sex']}"
)

model_7b, results_7b = logistic_model(formula)

print("\n=== Potassium × spironolactone: readmission ===")
print(results_7b)


# ============================================================
# VISUALIZATIONS
# ============================================================

sns.set_theme(style="whitegrid")


# eGFR stage
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="eGFR_stage",
    y=COLS["readmission_6m"],
    errorbar=("ci", 95)
)

plt.ylabel("6-month readmission proportion")
plt.xlabel("eGFR stage")
plt.title("6-month Readmission by eGFR Stage")
plt.tight_layout()
plt.show()


# Potassium
plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="K_group",
    y=COLS["death_6m"],
    errorbar=("ci", 95)
)

plt.ylabel("6-month mortality proportion")
plt.xlabel("Potassium")
plt.title("6-month Mortality by Potassium Level")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# Department
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x=COLS["department"],
    y=COLS["readmission_6m"],
    errorbar=("ci", 95)
)

plt.ylabel("6-month readmission proportion")
plt.xlabel("Discharge department")
plt.title("Readmission by Discharge Department")
plt.tight_layout()
plt.show()


# Age group
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="age_group",
    y=COLS["readmission_6m"],
    errorbar=("ci", 95)
)

plt.ylabel("6-month readmission proportion")
plt.xlabel("Age group")
plt.title("Readmission by Age Group")
plt.tight_layout()
plt.show()
```
